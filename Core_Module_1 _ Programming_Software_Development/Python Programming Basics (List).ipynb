# Accessing List Elements

Every element in a list has a unique position called an **index**.

Python uses **zero-based indexing**, which means indexing starts from **0**.

This allows us to access any element directly using its index.

# Visual Representation

Consider the following list:

```python
marks = [85, 90, 78, 95, 88]
```

```text
Element :   85    90    78    95    88
Index   :    0     1     2     3     4
```

- First element → Index `0`
- Last element → Index `4`

````python
marks = [85, 90, 78, 95, 88]

print(marks[0])
print(marks[1])
print(marks[2])
print(marks[3])
print(marks[4])
````

### Expected Output

```text
85
90
78
95
88
```

# Negative Indexing

Python also supports **negative indexing**.

Negative indexing starts counting from the **end** of the list.

```text
Element :   85    90    78    95    88
Index   :   -5    -4    -3    -2    -1
```

- `-1` → Last element
- `-2` → Second last element
- `-3` → Third last element

````python
marks = [85, 90, 78, 95, 88]

print(marks[-1])
print(marks[-2])
print(marks[-3])
````

### Expected Output

```text
88
95
78
```

# List Slicing

Slicing extracts a portion of a list.

### Syntax

```python
list_name[start : stop]
```

- **start** → Included
- **stop** → Excluded

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[0:3])
print(numbers[2:5])
print(numbers[1:4])
````

### Expected Output

```text
[10, 20, 30]

[30, 40, 50]

[20, 30, 40]
```

# Default Values in Slicing

If the start index is omitted, Python starts from index `0`.

If the stop index is omitted, Python continues until the end of the list.

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[:3])

print(numbers[3:])

print(numbers[:])
````

### Expected Output

```text
[10, 20, 30]

[40, 50, 60]

[10, 20, 30, 40, 50, 60]
```

# Slicing with Step

A third value called **step** controls how many elements Python skips.

### Syntax

```python
list_name[start : stop : step]
```

Default step is **1**.

````python
numbers = [10, 20, 30, 40, 50, 60, 70, 80]

print(numbers[::2])

print(numbers[1::2])

print(numbers[::3])
````

### Expected Output

```text
[10, 30, 50, 70]

[20, 40, 60, 80]

[10, 40, 70]
```

# Reversing a List

# Sorting Lists

Sorting arranges data in ascending or descending order.

Python provides two approaches:

- `sort()`
- `sorted()`

# sort()

The `sort()` method modifies the original list.

### Syntax

```python
list.sort()
```

````python
numbers = [45, 12, 98, 22, 67]

numbers.sort()

print(numbers)
````

````python
names = [
    "Rahul",
    "Aisha",
    "John",
    "Priya"
]

names.sort()

print(names)
````

````python
numbers = [10, 50, 20, 80, 30]

numbers.sort(reverse=True)

print(numbers)
````

The `sorted()` function returns a **new sorted list**.

The original list remains unchanged.

````python
numbers = [45, 12, 98, 22, 67]

new_numbers =sorted(numbers)
new_numbers.reverse()

print(numbers)

print(new_numbers)
````

# sort() vs sorted()

# Traversing a List

Traversing means visiting every element in a list.

This is one of the most important operations in programming.

````python
numbers = [10, 20, 30, 40]

for x in numbers:
    print(x)
````

This is the most commonly used approach.

Use it whenever you only need the values.

Python provides a cleaner way to access both index and value.

````python
students = [
    "Rahul",
    "Aisha",
    "John"
]

for index, student in enumerate(students):
    print(index, student)
````

`enumerate()` is widely used while processing datasets because row numbers are often important.

# Copying Lists

This is one of the most misunderstood topics for beginners.

Consider:

```python
a = [10,20,30]

b = a
```

Did Python create a new list?

Let's see.

````python
a = [10,20,30]

b = a

b.append(40)

print(a)

print(b)
````

Output:

```python
[10,20,30,40]

[10,20,30,40]
```

Both variables refer to the **same list**.

````python
a = [10,20,30]

b = a.copy()

b.append(40)

print(a)

print(b)
````

Output:

```python
a

[10,20,30]

b

[10,20,30,40]
```

Now both variables refer to **different lists**.

````python
a = [10,20,30]

b = a[:]

b.append(100)

print(a)

print(b)
````

# Ways to Copy Lists

| Method | Creates New List? |
|---------|-------------------|
| `=` | ❌ No |
| `copy()` | ✅ Yes |
| `list()` | ✅ Yes |
| `[:]` | ✅ Yes |

````python
training_loss = [0.91, 0.72, 0.54]

backup_loss = training_loss.copy()

training_loss.append(0.38)

print("Current :", training_loss)

print("Backup :", backup_loss)
````

Making a copy of data before modifying it is a common practice in machine learning pipelines.

It helps preserve the original dataset for comparison, validation, or rollback if needed.

# List Comprehension

List comprehension provides a concise way to create lists.

Instead of writing multiple lines of code, you can generate a list in a single expression.

### Syntax

```python
[expression for item in iterable]
```

Although optional for beginners, list comprehensions are widely used in Data Science and Machine Learning.

````python
numbers = []

for i in range(10):
    numbers.append(i)

print(numbers)
````

````python
numbers = [i for i in range(10)]

print(numbers)
````

Both programs produce the same result.

The second approach is shorter and more Pythonic.

````python
squares = [i**2 for i in range(10)]

print(squares)
````

````python
even_numbers = [i for i in range(20) if i % 2 == 0]

print(even_numbers)
````

````python
odd_numbers = [i for i in range(20) if i % 2 != 0]

print(odd_numbers)
````

# all()

Returns `True` only if **every** element is `True`.

````python
results = [True, True, True]

print(all(results))
````

````python
results = [True, False, True]

print(all(results))
````

````python
predictions = [1,0,1,1,0]

actual = [1,1,1,0,0]

comparison = [
    p == a
    for p, a in zip(predictions, actual)
]

print(comparison)
````

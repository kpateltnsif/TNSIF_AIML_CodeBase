# What is a Function?

A **function** is a reusable block of code that performs a specific task.

Instead of writing the same code repeatedly, we write it once inside a function and call it whenever needed.

Think of a function as a machine:

```text
Input
   │
   ▼
+-------------+
|  Function   |
+-------------+
   │
   ▼
Output
```

Functions help make programs:

- Modular
- Reusable
- Easy to read
- Easy to maintain

# Why Do We Need Functions?

Without functions, the same code must be written repeatedly.

Example (without functions):

```python
print("Welcome Rahul")
print("Welcome Aisha")
print("Welcome John")
```

Using a function:

```python
def welcome(name):
    print(f"Welcome {name}")

welcome("Rahul")
welcome("Aisha")
welcome("John")
```

One function can be reused many times.

# Advantages of Functions

- Reduce code duplication
- Improve readability
- Simplify debugging
- Promote code reuse
- Make large programs easier to manage
- Improve modularity

# Function Syntax

Functions are defined using the `def` keyword.

### Syntax

```python
def function_name():
    statements
```

To execute a function, it must be **called**.

````python
def greet():
    print("Welcome to Python")

greet()
greet()
````

# Defining vs Calling a Function

There are two steps:

## 1. Define the Function

```python
def greet():
    print("Hello")
```

## 2. Call the Function

```python
greet()
```

A function does nothing until it is called.

````python
def greet():
    print("Welcome!")

greet()
greet()
greet()
````

# Parameters

Parameters allow data to be passed into a function.

### Syntax

```python
def function(parameter):
    statements
```

The parameter behaves like a local variable inside the function.

````python
def greet(name):
    print(f"Welcome {name}")

greet("Rahul")
greet("Aisha")
greet("John")
````

````python
def display_marks(marks):
    print("Marks:", marks)

display_marks(91)
display_marks(88)
display_marks(95)
````

A single function can work with different values depending on the argument passed during the function call.

````python
def student(name, age):

    print("Name:", name)

    print("Age :", age)

student("Rahul", 21)
````

````python
def result(name, marks):

    print(f"{name} scored {marks}")

result("Aisha", 92)
````

# Parameters vs Arguments

These terms are often confused.

### Parameters

Variables defined in the function definition.

```python
def greet(name):
```

Here,

```python
name
```

is a **parameter**.

---

### Arguments

Actual values passed during the function call.

```python
greet("Rahul")
```

Here,

```python
"Rahul"
```

is an **argument**.

# The `return` Statement

Functions can return values back to the caller.

### Syntax

```python
return value
```

Returned values can be stored in variables or used directly.

````python
def square(number):

    return number ** 2

result = square(5)

print(result)
````

````python
def add(a, b):

    return a + b

answer = add(12, 18)

print(answer)
````

# `print()` vs `return`

`print()`

- Displays a value on the screen.
- Does not send it back to the caller.

`return`

- Sends the value back to the caller.
- Can be stored, reused, or passed to another function.

In most real-world programs, **returning values is preferred over printing them**.

````python
def calculate_accuracy(correct, total):

    return (correct / total) * 100

accuracy = calculate_accuracy(94, 100)

print("Accuracy:", accuracy)
````

````python
def dataset_info(rows, columns):

    print("Rows    :", rows)

    print("Columns :", columns)

dataset_info(5000, 18)
````

````python
def display_feature(feature):

    print("Feature:", feature)

display_feature("Age")
display_feature("Salary")
display_feature("Experience")
````

````python
def evaluate_model(accuracy):

    if accuracy >= 90:
        return "Accepted"

    return "Rejected"

print(evaluate_model(94))
print(evaluate_model(82))
````

# Function Arguments

Python provides several ways to pass arguments to functions.

In this section, you'll learn:

- Positional Arguments
- Keyword Arguments
- Default Arguments
- Variable-Length Arguments (`*args`)
- Keyword Variable-Length Arguments (`**kwargs`)
- Local Variables
- Global Variables

These concepts are widely used in Python libraries, APIs, Machine Learning, and Data Science.

# Positional Arguments

Arguments are assigned to parameters based on their position.

### Syntax

```python
function(argument1, argument2)
```

The order of arguments is important.

````python
def student(name, age):
    print("Name :", name)
    print("Age  :", age)

student("Rahul", 21)
````

````python
def subtract(a, b):
    print(a - b)

subtract(20, 5)
subtract(5, 20)
````

Changing the order of positional arguments changes the result.

# Keyword Arguments

Keyword arguments explicitly specify which parameter receives each value.

### Syntax

```python
function(parameter=value)
```

The order of keyword arguments does not matter.

````python
def student(name, age):
    print("Name :", name)
    print("Age  :", age)

student(age=21, name="Rahul")
````

````python
def model(name, accuracy):
    print("Model    :", name)
    print("Accuracy :", accuracy)

model(accuracy=95.4, name="Random Forest")
````

Keyword arguments improve readability, especially when a function has many parameters.

# Default Arguments

A parameter can have a default value.

If no argument is passed, the default value is used.

### Syntax

```python
def function(parameter=default_value):
```

````python
def greet(name="Student"):
    print(f"Welcome {name}")

greet()
greet("Rahul")
````

````python
def train_model(epochs=10):
    print(f"Training for {epochs} epochs")

train_model()
train_model(25)
````

Default arguments make functions more flexible by allowing optional parameters.

#Multiple Default Arguments

````python
def employee(name, department="IT", location="Mumbai"):
    print("Name       :", name)
    print("Department :", department)
    print("Location   :", location)

employee("Rahul")
employee("Aisha", "AI")
employee("John", "HR", "Pune")
````

# Variable-Length Arguments (`*args`)

Sometimes we don't know how many arguments will be passed.

Use `*args` to accept any number of positional arguments.

### Syntax

```python
def function(*args):
```

Inside the function, `args` is stored as a **tuple**.

````python
def numbers(*args):
    print(args)

numbers(10, 20)
numbers(10, 20, 30, 40)
````

````python
def total(*numbers):
    print(sum(numbers))

total(10, 20, 30)
total(5, 10, 15, 20, 25)
````

````python
def display(*items):
    for item in items:
        print(item)

display("Python", "Machine Learning", "Deep Learning")
````

## 🧠 AI Example

A function that calculates the average of any number of accuracy values.

````python
def average_accuracy(*accuracies):
    return sum(accuracies) / len(accuracies)

print(average_accuracy(91, 93, 95))
print(average_accuracy(88, 89, 90, 91, 92))
````

# Keyword Variable-Length Arguments (`**kwargs`)

`**kwargs` accepts any number of keyword arguments.

Inside the function, they are stored as a **dictionary**.

### Syntax

```python
def function(**kwargs):
```

````python
def student(**details):
    print(details)

student(name="Rahul", age=21, city="Nashik")
````

````python
def student(**details):
    for key, value in details.items():
        print(key, ":", value)

student(
    name="Rahul",
    age=21,
    city="Nashik"
)
````

## 🧠 AI Example

Store machine learning model information dynamically.

````python
def model(**config):
    for key, value in config.items():
        print(key, ":", value)

model(
    algorithm="Random Forest",
    trees=100,
    max_depth=12,
    accuracy=95.4
)
````

# Local Variables

A variable created inside a function is called a **local variable**.

It exists only while the function is executing.

````python
def display():
  message = "Hello"

  print(message)

display()

print(message)
````

Trying to access a local variable outside the function results in a `NameError`.

# Global Variables

A variable declared outside every function is called a **global variable**.

Global variables can be accessed from inside functions.

````python
course = "Python"

def display():
    print(course)

display()
````

````python
model_name = "Random Forest"

def show_model():
    print("Current Model:", model_name)

show_model()
````

# The `global` Keyword

Normally, variables created inside functions are local.

To modify a global variable inside a function, use the `global` keyword.

````python
count = 1

def increment():
    global count
    count += 1

increment()
increment()

print(count)
````

````python
def evaluate_model(name, accuracy=90):
    print(f"Model: {name}")
    print(f"Accuracy: {accuracy}")

evaluate_model("Decision Tree")
evaluate_model("Random Forest", 95.2)
````

````python
def display_features(*features):
    for feature in features:
        print(feature)

display_features(
    "Age",
    "Salary",
    "Experience",
    "Education"
)
````

````python
def model_config(**config):
    for key, value in config.items():
        print(key, ":", value)

model_config(
    learning_rate=0.01,
    epochs=100,
    batch_size=32
)
````

# Lambda Functions

A **Lambda Function** is an anonymous (unnamed) function.

It is generally used for small, simple operations.

### Syntax

```python
lambda parameters: expression
```

Equivalent function:

```python
def square(x):
    return x * x
```

Lambda version:

```python
lambda x: x * x
```

````python
square = lambda x: x ** 2

print(square(5))
````

````python
add = lambda a, b: a + b

print(add(10, 20))
````

````python
maximum = lambda a, b: a if a > b else b

print(maximum(15, 28))
````

Use lambda functions only for short, simple operations.

For complex logic, prefer normal functions using `def`.

````python
accuracy_check = lambda accuracy: "Accepted" if accuracy >= 90 else "Rejected"

print(accuracy_check(95))

print(accuracy_check(84))
````

# Recursion

A function that calls itself is called a **Recursive Function**.

Every recursive function must have:

- Base Case
- Recursive Case

Without a base case, recursion continues indefinitely.

## Recursion Flow

```text
Function

↓

Calls Itself

↓

Calls Itself Again

↓

...

↓

Base Case Reached

↓

Returns Back
```

````python
def countdown(number):

    if number == 0:
        return

    print(number)

    countdown(number - 1)

countdown(5)
````

````python
def factorial(number):

    if number == 1:
        return 1

    return number * factorial(number - 1)

print(factorial(5))
````

Execution:

```text
factorial(5)

↓

5 × factorial(4)

↓

5 × 4 × factorial(3)

↓

...

↓

5 × 4 × 3 × 2 × 1
```

````python
def total(number):

    if number == 1:
        return 1

    return number + total(number - 1)

print(total(10))
````

Although recursion is elegant, many problems can be solved more efficiently using loops.

Use recursion when it naturally models the problem, such as tree traversal or divide-and-conquer algorithms.

# Docstrings

A **Docstring** is a string written immediately below the function definition.

It explains:

- What the function does
- Parameters
- Return value

Docstrings improve readability and documentation.

````python
def square(number):
    """
    Returns the square of a number.
    """

    return number ** 2

print(square(5))
````

````python
def greet(name):
    """
    Displays a welcome message.
    """

    print(f"Welcome {name}")

help(greet)
````

# Type Hints

Type hints specify the expected data types of parameters and return values.

They improve readability and help code editors detect errors.

Python does **not** enforce type hints during execution.

````python
def add(a: int, b: int) -> int:
    return a + b

print(add(10, 20.0))
````

````python
def calculate_accuracy(correct: int, total: int) -> float:
    return (correct / total) * 100

print(calculate_accuracy(94, 100))
````

Type hints improve code quality, especially in large AI and Data Science projects.

# First-Class Functions

In Python, functions are treated as objects.

A function can be:

- Assigned to a variable
- Passed as an argument
- Returned from another function

````python
def greet():
    print("Hello")

message = greet

message()
````

# Higher-Order Functions

A higher-order function either:

- Accepts another function as an argument, or
- Returns a function.

````python
def greet():
    print("Welcome")

def execute(function):
    function()

execute(greet)
````

Higher-order functions are widely used in Python libraries.

Examples include:

- `map()`
- `filter()`
- `sorted()`

These will be covered in the Built-in Functions notebook.

````python
predict = lambda probability: "Spam" if probability >= 0.5 else "Not Spam"

print(predict(0.82))
print(predict(0.34))
````

````python
def train_epoch(epoch: int) -> None:
    """
    Simulates one training epoch.
    """

    print(f"Training Epoch {epoch}")

for epoch in range(1, 6):
    train_epoch(epoch)
````

````python
def feature_summary(**features):
    """
    Displays feature information.
    """

    for feature, value in features.items():
        print(feature, ":", value)

feature_summary(
    Age="Numeric",
    Salary="Numeric",
    Purchased="Target"
)
````

### ❌ Mistake 1

Using lambda functions for long and complex logic.

Prefer normal functions (`def`) for readability.

---

### ❌ Mistake 2

Writing recursive functions without a base case.

This causes infinite recursion and eventually raises a `RecursionError`.

---

### ❌ Mistake 3

Confusing `return` with `print()` inside recursive functions.

Recursive functions almost always need `return`.

---

### ❌ Mistake 4

Assuming type hints enforce data types.

They improve readability but do not prevent incorrect types during execution.

# What is a Loop?

A loop allows a block of code to execute **multiple times** without writing the same code repeatedly.

Instead of writing:

```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```

We can write:

```python
for i in range(5):
    print("Hello")
```

Loops make programs:

- Shorter
- Cleaner
- Easier to maintain
- More efficient

# Why Do We Need Loops?

Loops are used whenever a task needs to be repeated.

Examples:

- Display numbers from 1 to 100
- Process every student record
- Calculate the average of marks
- Traverse a dataset
- Read every line in a file
- Train a model for multiple epochs
- Process images one by one

Without loops, these tasks would require writing repetitive code.

# Types of Loops in Python

Python provides two looping statements:

## 1. `for` Loop

Used when the number of iterations is known or when iterating over a collection.

Examples:

- Lists
- Strings
- Tuples
- Dictionaries
- Sets
- `range()`

---

## 2. `while` Loop

Used when the number of iterations is **not known in advance**.

The loop continues until a condition becomes `False`.

# The `for` Loop

The `for` loop iterates over each item in a sequence.

### Syntax

```python
for variable in sequence:
    statement
```

The sequence can be:

- String
- List
- Tuple
- Set
- Dictionary
- `range()`

````python
for i in range(5):
    print("Hello")
````

````python
for number in range(1, 6):
    print(number)
````

# Execution Flow

```text
Start

↓

Take first value

↓

Execute loop body

↓

Take next value

↓

Execute again

↓

Continue until sequence ends

↓

Exit loop
```

# The `range()` Function

The `range()` function generates a sequence of numbers.

It is commonly used with `for` loops.

### Syntax

```python
range(stop)
```

```python
range(start, stop)
```

```python
range(start, stop, step)
```

> **Note:** The `stop` value is **not included** in the generated sequence.

````python
for i in range(5):
    print(i)
````

````python
for i in range(2, 8):
    print(i)
````

````python
for i in range(0, 11, 2):
    print(i)
````

````python
for i in range(10, 0, -1):
    print(i)
````

| `range(2, 6)` | 2 3 4 5 |
| `range(0, 10, 2)` | 0 2 4 6 8 |
| `range(10, 0, -2)` | 10 8 6 4 2 |

# Traversing Data Structures

# Traversing a String

A string is a sequence of characters.

A `for` loop processes one character at a time.

````python
text = "Python"

for character in text:
    print(character)
````

````python
course = "Machine Learning"

for character in course:
    print(character)
````

# Traversing a List

Lists are the most frequently used data structure in Python.

````python
students = [
    "Rahul",
    "Aisha",
    "John",
    "Priya"
]

for student in students:
    print(student)
````

````python
marks = [85, 92, 76, 88]

for mark in marks:
    print(mark)
````

## 🧠 AI Example

A dataset often contains thousands of records.

````python
coordinates = (15, 28, 35, 42)

for value in coordinates:
    print(value)
````

# Traversing Dictionaries

Dictionaries can be traversed in several ways.

- Keys
- Values
- Key-Value Pairs

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

for key in student:
    print(key)
````

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

for value in student.values():
    print(value)
````

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

for key, value in student.items():
    print(key, ":", value)
````

When both keys and values are required,

prefer using:

```python
items()
```

This is the most common approach in professional Python code.

# The `while` Loop

Unlike the `for` loop,

a `while` loop executes **as long as a condition remains True**.

### Syntax

```python
while condition:
    statements
```

The condition is checked before every iteration.

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

# Execution Flow

```text
Condition

↓

True?

↓

Execute Loop

↓

Update Condition

↓

Check Again

↓

Condition becomes False

↓

Exit Loop
```

````python
count = 5

while count > 0:
    print(count)
    count -= 1

print("Blast Off!")
````

````python
epoch = 1

while epoch <= 5:
    print("Training Epoch", epoch)
    epoch += 1
````

````python
attempt = 1

while attempt <= 3:
    print("Prediction Attempt", attempt)
    attempt += 1
````

# Infinite Loops

Example:

```python
while True:
    print("Hello")
```

This loop runs forever unless interrupted.

Infinite loops are useful in:

- Game loops
- Servers
- Chatbots
- Monitoring systems

However, they should be used carefully.

# `for` vs `while`

| `for` Loop | `while` Loop |
|------------|--------------|
| Used when the number of iterations is known | Used when the number of iterations is unknown |
| Iterates over a sequence | Depends on a condition |
| Less prone to infinite loops | Can easily become infinite if the condition is not updated |
| Preferred for collections | Preferred for condition-based repetition |

````python
dataset = [
    "Record 1",
    "Record 2",
    "Record 3",
    "Record 4"
]

for record in dataset:
    print("Processing", record)
````

````python
accuracy_history = [
    82.5,
    86.7,
    90.2,
    93.4,
    95.1
]

for accuracy in accuracy_history:
    print("Accuracy:", accuracy)
````

````python
feature_importance = {
    "Age": 0.42,
    "Salary": 0.31,
    "Experience": 0.18,
    "Education": 0.09
}

for feature, score in feature_importance.items():
    print(feature, "->", score)
````

### ❌ Mistake 1

Forgetting to update the loop variable in a `while` loop.

```python

    print(count)
```

This creates an **infinite loop**.

---

### ❌ Mistake 2

Using a `while` loop when a `for` loop is simpler.

If you're iterating over a list, string, tuple, or dictionary, prefer a `for` loop.

---

### ❌ Mistake 3

Expecting sets to maintain insertion order during iteration.

Sets are unordered.

---

### ❌ Mistake 4

Using dictionary indexing.

Incorrect:

```python
student[0]
```

Correct:

```python
student["Name"]
```

# Nested Loops

A loop inside another loop is called a **Nested Loop**.

### Syntax

```python
for outer in sequence1:
    for inner in sequence2:
        statements
```

For every iteration of the outer loop, the inner loop executes completely.

````python
for i in range(1, 4):
    for j in range(1, 4):
      for x in range(1, 4):
          print(i, j , x)
````

````python
for row in range(1, 4):
    for column in range(1, 4):
        print(f"({row}, {column})", end=" ")
    print()
````

### Execution Flow

For:

```python
for i in range(2):
    for j in range(3):
```

Execution order:

```text
i = 0

    j = 0

    j = 1

    j = 2

i = 1

    j = 0

    j = 1

    j = 2
```

````python
dataset = [
    [21, 35000, 0],
    [25, 52000, 1],
    [30, 68000, 1]
]

for row in dataset:
    for value in row:
        print(value, end=" ")
    print()
````

# `break`

The `break` statement immediately terminates the nearest loop.

Execution continues with the first statement after the loop.

### Syntax

```python
break
```

````python
for number in range(1, 11):

    if number == 6:
        break

    print(number)
````


Output:

```text
1
2
3
4
5
```

The loop stops when `number` becomes `6`.

````python
count = 1

while True:

    if count == 6:
        break

    print(count)

    count += 1
````

## 🧠 AI Example

Stop training once the desired accuracy is achieved.

````python
accuracy_history = [82, 87, 91, 94, 97, 98]

for accuracy in accuracy_history:

    print("Accuracy:", accuracy)

    if accuracy >= 95:
        print("Target Accuracy Achieved")
        break
````

# `continue`

The `continue` statement skips the current iteration and moves to the next iteration.

````python
for number in range(1, 11):

    if number == 5:
        continue

    print(number)
````


Output:

```text
1
2
3
4
6
7
8
9
10
```

The value `5` is skipped.

````python
for number in range(1, 11):

    if number % 2 == 0:
        continue

    print(number)
````

## 🧠 AI Example

Skip invalid records while processing a dataset.

````python
dataset = [12, None, 18, None, 25]

for record in dataset:

    if record is None:
        continue

    print("Processing:", record)
````

# `pass` in Loops

The `pass` statement performs no action.

It is commonly used while writing incomplete code.

````python
for number in range(5):

    if number == 2:
        pass

    print(number)
````

Unlike `break` and `continue`, `pass` does **not** affect loop execution.

It simply acts as a placeholder.

# `else` with Loops

Python allows an `else` block after both `for` and `while` loops.

The `else` block executes **only if the loop finishes normally**.

If the loop is terminated using `break`, the `else` block is skipped.

````python
for number in range(1, 6):
    print(number)

else:
    print("Loop Completed")
````

````python
for number in range(1, 6):

    if number == 3:
        break

    print(number)

else:
    print("Loop Completed")
````

Since the loop ended using `break`, the `else` block does **not** execute.

# `enumerate()`

Sometimes we need both:

- Index
- Value

Instead of manually maintaining a counter, Python provides `enumerate()`.

### Syntax

```python
enumerate(iterable)
```

````python
students = [
    "Rahul",
    "Aisha",
    "John"
]

for index, student in enumerate(students):
    print(index, student)
````

````python
students = [
    "Rahul",
    "Aisha",
    "John"
]

for index, student in enumerate(students, start=1):
    print(index, student)
````

## 🧠 AI Example

Display feature numbers while processing feature names.

````python
features = [
    "Age",
    "Salary",
    "Experience",
    "Education"
]

for index, feature in enumerate(features, start=1):
    print(f"Feature {index}: {feature}")
````

````python
dataset = [
    [21, 35000, 0],
    [24, 42000, 1],
    [30, 61000, 1]
]

for row_number, row in enumerate(dataset, start=1):

    print(f"Row {row_number}")

    for value in row:
        print(value)

    print("-" * 20)
````

````python
accuracies = [81, 85, 89, 94, 97]

for accuracy in accuracies:

    if accuracy >= 95:
        print("Suitable Model Found")
        break
````

````python
dataset = [12, 15, None, 18, None, 22]

for value in dataset:

    if value is None:
        continue

    print(value)
````

````python
feature_names = [
    "Age",
    "Salary",
    "Experience",
    "Education",
    "Purchased"
]

for index, feature in enumerate(feature_names, start=1):
    print(index, feature)
````

### ❌ Mistake 1

Using `break` instead of `continue`.

- `break` → exits the loop completely.
- `continue` → skips only the current iteration.

---

### ❌ Mistake 2

Expecting the `else` block to execute after `break`.

It executes **only if the loop finishes normally**.

---

### ❌ Mistake 3

Using a manual counter instead of `enumerate()`.

Prefer:

```python
for index, value in enumerate(data):
```

---

### ❌ Mistake 4

Confusing `pass` with `continue`.

- `pass` does nothing.
- `continue` skips the remaining statements of the current iteration.

# Quick Recap

✔ `else` executes only when the loop finishes without `break`.

✔ `enumerate()` provides both index and value during iteration.

# Pandas — Part 1
## Introduction, Series and DataFrames

Pandas is a Python library used for working with structured and tabular data.

It is widely used in:

- Data Science
- Data Analysis
- Machine Learning
- Data Cleaning
- Exploratory Data Analysis (EDA)

Examples of data that can be handled using Pandas:

- Student records
- Customer data
- Sales data
- Employee information
- House prices
- Financial datasets
- Machine Learning datasets

Pandas mainly provides two important data structures:

1. Series
2. DataFrame

`````python
import numpy as np
import pandas as pd
`````

## 1. Pandas Series

A **Series** is a one-dimensional labeled data structure.

It can be thought of as:

NumPy Array + Labels

A Series contains:

- Values
- Index labels

By default, Pandas assigns indexes starting from `0`.

`````python
marks = pd.Series([75, 82, 91, 68, 88])

print(marks)
`````

## 2. Accessing Series Elements

A Series can be accessed using its index.

For example:

- Index `0` → first value
- Index `1` → second value
- Index `2` → third value

`````python
marks = pd.Series([75, 82, 91, 68, 88])

print(marks[0])
print(marks[2])
print(marks[4])
`````

## 3. Custom Index

We can provide our own labels instead of using the default numerical index.

This is useful when the labels have meaning, such as:

- Student names
- Product IDs
- Dates
- Employee IDs

`````python
marks = pd.Series(
    [75, 82, 91, 68, 88],
    index=["A", "B", "C", "D", "E"]
)

print(marks)
`````

## 4. Accessing Data Using Custom Index

When a Series has custom indexes, we can use those labels to access values.

`````python
print(marks["A"])
print(marks["C"])
print(marks["E"])
`````

## 5. Creating a Series from a Dictionary

A Python dictionary can be directly converted into a Pandas Series.

The:

- Dictionary keys become the Series indexes.
- Dictionary values become the Series values.

`````python
marks = {
    "Rahul": 75,
    "Amit": 82,
    "Priya": 91,
    "Neha": 68
}

marks_series = pd.Series(marks)

print(marks_series)
`````

## 6. Series Properties

Important Series properties include:

- `shape` → dimensions of the Series
- `size` → total number of elements
- `ndim` → number of dimensions
- `dtype` → data type of the values

`````python
marks = pd.Series([75, 82, 91, 68, 88])

print("Shape:", marks.shape)
print("Size:", marks.size)
print("Dimensions:", marks.ndim)
print("Data Type:", marks.dtype)
`````

## 7. Mathematical Operations on Series

Pandas Series support vectorized mathematical operations.

Operations can be performed directly on the entire Series without writing a loop.

`````python
marks = pd.Series([75, 82, 91, 68, 88])

print("Original:")
print(marks)

print("\nAdded 5:")
print(marks + 5)

print("\nMultiplied by 2:")
print(marks * 2)

print("\nDivided by 2:")
print(marks / 2)
`````

## 8. Statistical Operations on Series

Pandas provides built-in methods for common statistical calculations.

Important methods include:

- `mean()`
- `median()`
- `min()`
- `max()`
- `sum()`
- `std()`
- `var()`

These are frequently used during Data Analysis and EDA.

`````python
marks = pd.Series([75, 82, 91, 68, 88])

print("Mean:", marks.mean())
print("Median:", marks.median())
print("Minimum:", marks.min())
print("Maximum:", marks.max())
print("Sum:", marks.sum())
print("Standard Deviation:", marks.std())
print("Variance:", marks.var())
`````

## 9. Filtering a Series

A Series can be filtered using a condition.

The general pattern is:

```python
series[condition]

`````python


marks = pd.Series([45, 67, 89, 32, 76, 91, 55, 84])

high_scores = marks[marks > 70]

print(high_scores)
`````

# DataFrame

A **DataFrame** is a two-dimensional labeled data structure.

It can be thought of as a table containing:

- Rows
- Columns
- Index
- Values

For Data Science, the DataFrame is the most commonly used Pandas structure.

Example:

| Name | Age | Marks |
|---|---:|---:|
| Rahul | 21 | 78 |
| Amit | 22 | 85 |
| Priya | 20 | 92 |
| Neha | 21 | 74 |

Here:

- Each column represents a variable or feature.
- Each row represents an observation.

`````python
data = {
    "Name": ["Rahul", "Amit", "Priya", "Neha"],
    "Age": [21, 22, 20, 21],
    "Marks": [78, 85, 92, 74]
}

df = pd.DataFrame(data)

print(df)
`````

`````python
import numpy as np
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Age": [20, 21, 22, 20, 23, 21],
    "Study_Hours": [4, 6, 8, 3, 7, 5],
    "Attendance": [75, 82, 91, 65, 88, 79],
    "Marks": [62, 74, 89, 55, 84, 71]
}

df = pd.DataFrame(data)

df
`````

## 1. Single-Column Selection

A single column can be selected using its column name.

The syntax is:

`df["column_name"]`

Selecting a single column returns a **Pandas Series**.

`````python
marks = df["Marks"]

print(marks)
print(type(marks))
`````

## 2. Multiple-Column Selection

Multiple columns can be selected by passing a list of column names.

The result is a **Pandas DataFrame**.

This is useful when we only need a subset of the available variables.

`````python
selected_data = df[
    ["Student", "Age", "Marks"]
]

selected_data
`````

## 3. `loc[]`

`loc[]` performs **label-based selection**.

The general syntax is:

`df.loc[row_label, column_label]`

It can be used to select:

- A specific row
- A specific value
- Multiple rows
- Specific rows and columns

`````python
# Select the row with index label 0

df.loc[0]
`````

`````python
# Select a specific value

df.loc[2, "Marks"]
`````

`````python
# Select rows from index 1 through 3

df.loc[1:3]
`````

`````python
# Select specific rows and columns

df.loc[
    1:3,
    ["Student", "Marks"]
]
`````

## 4. `iloc[]`

`iloc[]` performs **position-based selection**.

The general syntax is:

`df.iloc[row_position, column_position]`

Positions start from `0`.

The important difference is:

| Method | Based On |
|---|---|
| `loc[]` | Labels |
| `iloc[]` | Integer positions |

`````python
# Select the first row

df.iloc[0]
`````

`````python
# Select the value at row position 2 and column position 4

df.iloc[2, 4]
`````

`````python
# Select rows at positions 1, 2 and 3

df.iloc[1:4]
`````

`````python
# Select rows 1 to 3 and columns 0 and 4

df.iloc[
    1:4,
    [0, 4]
]
`````

## 5. Boolean Filtering

Boolean filtering allows us to select only rows that satisfy a condition.

The general pattern is:

`df[condition]`

For example, we can select students whose marks are greater than `75`.

`````python
high_scorers = df[
    df["Marks"] > 75
]

high_scorers
`````

## 6. Multiple Conditions Using `&`

The `&` operator represents **AND**.

Both conditions must be true.

For example, select students who:

- Study for more than 5 hours
- AND have attendance greater than 80

`````python
result = df[
    (df["Study_Hours"] > 5) &
    (df["Attendance"] > 80)
]

result
`````

## 7. Multiple Conditions Using `|`

The `|` operator represents **OR**.

At least one of the conditions must be true.

For example, select students who:

- Have marks greater than 85
- OR have attendance greater than 90

`````python
result = df[
    (df["Marks"] > 85) |
    (df["Attendance"] > 90)
]

result
`````

## 8. `isin()`

`isin()` is useful when we want to check whether values belong to a specific list.

For example, we can select students whose age is either:

- 20
- 23

`````python
result = df[
    df["Age"].isin([20, 23])
]

result
`````

## 9. `between()`

`between()` is used to select values within a specified range.

For example, we can select students whose marks are between `70` and `85`.

By default, the boundary values are included.

`````python
result = df[
    df["Marks"].between(70, 85)
]

result
`````

# 10. Basic `X` and `y` Preparation

In supervised Machine Learning, we generally separate the dataset into:

- `X` → Feature Matrix
- `y` → Target Variable

Suppose we want to predict student marks using:

- Age
- Study Hours
- Attendance

Then these three columns become our features.

`Marks` becomes the target.

The structure is:

X → Input Features

y → Target

`````python
X = df[
    ["Age", "Study_Hours", "Attendance"]
]

y = df["Marks"]

print("Feature Matrix X:")
print(X)

print("\nTarget y:")
print(y)

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)
`````

# Common Mistakes

## 1. Using `and` Instead of `&`

For Pandas Series, use `&` for element-wise AND.

Do not use Python's `and`.

## 2. Using `or` Instead of `|`

For Pandas Series, use `|` for element-wise OR.

## 3. Forgetting Parentheses

Each condition should be enclosed in parentheses when using `&` or `|`.

## 4. Confusing `loc[]` and `iloc[]`

Remember:

`loc[]` → Labels

`iloc[]` → Integer positions

## 5. Forgetting Double Brackets

Single column:

`df["Marks"]`

Multiple columns:

`df[["Marks", "Age"]]`

## 6. Assuming Filtering Changes the Original DataFrame

A filtering operation normally creates a result.

It does not automatically modify the original DataFrame.

## 7. Ignoring Missing Values

Real-world datasets can contain missing values.

For example, Titanic contains missing values in columns such as `Age`.

Missing-value handling will be covered separately.

`````python
import numpy as np
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Age": [20, 21, 22, 20, 23, 21],
    "Study_Hours": [4, 6, 8, 3, 7, 5],
    "Attendance": [75, 82, 91, 65, 88, 79],
    "Marks": [62, 74, 89, 55, 84, 71]
}

df = pd.DataFrame(data)

df
`````

## 1. Creating a New Column

A new column can be created by assigning a value to a new column name.

The new column can contain:

- A constant value
- Calculated values
- Boolean values
- Values derived from existing columns

Creating new columns is one of the most common operations in Data Science.

For example, we can create a `Passed` column based on whether a student's marks are at least 40.

The result will be:

- `1` → Passed
- `0` → Failed

This is also an example of creating a feature that can later be used in Machine Learning.

`````python
df["Passed"] = (df["Marks"] >= 40).astype(int)

df
`````

## 2. Creating a Calculated Column

We can create a new column by performing a calculation on an existing column.

Suppose `Study_Hours` represents the number of hours a student studies.

We can convert it into minutes:

`Study_Minutes = Study_Hours × 60`

This is useful when we want the same information represented in a different form.

`````python
df["Study_Minutes"] = df["Study_Hours"] * 60

df
`````

## 3. Creating a Column Using Multiple Columns

A new feature can be created using multiple existing columns.

Suppose we want to calculate a performance score using:

- Marks → 70%
- Attendance → 30%

The formula is:

`Performance_Score = Marks × 0.70 + Attendance × 0.30`

This is an example of **Feature Engineering**.

Feature Engineering means creating useful features from existing data.

`````python
df["Performance_Score"] = (
    df["Marks"] * 0.70 +
    df["Attendance"] * 0.30
)

df
`````

## 4. Modifying an Existing Column

An existing column can be transformed by assigning a new value to it.

For example, marks are currently represented from 0 to 100.

We can convert them into a decimal representation between 0 and 1.

Important:

When we assign the result back to the original column, the original values are replaced.

Therefore, when we want to preserve the original data, it is usually better to create a new column.

`````python
df["Marks"] = df["Marks"] / 100

df
`````

## Restoring the Original Marks

For the remaining examples, we want `Marks` to remain on a 0–100 scale.

We can restore the original values so that the following examples are easier to understand.

`````python
df["Marks"] = df["Marks"] * 100

df
`````

## 5. Creating Categorical Features

Sometimes numerical values are easier to understand when converted into categories.

For example:

- Marks >= 80 → Excellent
- Marks >= 60 → Good
- Marks >= 40 → Average
- Marks < 40 → Needs Improvement

This type of transformation is useful in:

- Data Analysis
- Reporting
- Exploratory Data Analysis
- Feature Engineering

We can use NumPy's `select()` to create the categories.

`````python
conditions = [
    df["Marks"] >= 80,
    df["Marks"] >= 60,
    df["Marks"] >= 40
]

choices = [
    "Excellent",
    "Good",
    "Average"
]

df["Performance_Level"] = np.select(
    conditions,
    choices,
    default="Needs Improvement"
)

df
`````

## 6. Using `assign()`

`assign()` provides another way to create new columns.

It returns a new DataFrame containing the added or modified columns.

It is particularly useful when performing several transformations as part of a data-processing workflow.

One advantage is that it can make transformation pipelines easier to read.

`````python
df = df.assign(
    Study_Days=df["Study_Hours"] / 2
)

df
`````

## 7. Renaming Columns

Column names can be changed using `rename()`.

Renaming is useful when:

- Names are unclear
- Names are inconsistent
- Names contain unnecessary spaces
- We want standardized naming
- We want names that are easier to use in code

For example, we can change:

`Student` → `Student_ID`

`Marks` → `Final_Marks`

`````python
df = df.rename(
    columns={
        "Student": "Student_ID",
        "Marks": "Final_Marks"
    }
)

df
`````

## 8. Dropping Columns

Unnecessary columns can be removed using `drop()`.

A column may be removed because:

- It is irrelevant
- It contains duplicate information
- It is no longer required
- It is not useful for the current analysis

We should always understand a column before removing it.

Here, we will remove the temporary `Study_Days` column.

`````python
df = df.drop(
    columns=["Study_Days"]
)

df
`````

## 9. Dropping Rows

Rows can also be removed using `drop()`.

We specify the index of the row that should be removed.

For demonstration, we will create a separate DataFrame so that our original dataset remains unchanged.

`````python
df_without_first_row = df.drop(index=0)

df_without_first_row
`````

## 10. Inserting a Column

`insert()` allows us to create a new column at a specific position.

The general structure is:

`df.insert(position, column_name, values)`

The position starts from `0`.

This is useful when the order of columns matters for readability or reporting.

`````python
df.insert(
    2,
    "Study_Minutes_Copy",
    df["Study_Minutes"]
)

df
`````

## Removing the Demonstration Column

The previous column was created only to demonstrate `insert()`.

We can remove it using `drop()`.

`````python
df = df.drop(
    columns=["Study_Minutes_Copy"]
)

df
`````

## 11. Reordering Columns

We can rearrange the columns into a specific order.

This can make a dataset easier to read and can be useful when preparing data for further processing.

We simply create a list containing the desired column order and use it to reorganize the DataFrame.

`````python
column_order = [
    "Student_ID",
    "Age",
    "Study_Hours",
    "Study_Minutes",
    "Attendance",
    "Final_Marks",
    "Passed",
    "Performance_Score",
    "Performance_Level"
]

df = df[column_order]

df
`````

# Data Science Example
## Creating an Overall Performance Feature

Feature Engineering often combines multiple variables into a new feature.

Suppose we create an overall score using:

- Final Marks → 60%
- Attendance → 30%
- Study Hours → 10%

Because Study Hours can be on a different numerical scale, we will use a simple weighted formula for demonstration.

This is not necessarily the best statistical formula for a real model.

The important concept is how existing columns can be transformed into a new feature.

`````python
df["Overall_Score"] = (
    df["Final_Marks"] * 0.60 +
    df["Attendance"] * 0.30 +
    df["Study_Hours"] * 2
)

df
`````

# AI / Machine Learning Example
## Creating a Classification Target

Suppose we want to classify students into:

- High Performer
- Not High Performer

We can define a simple rule:

`Overall_Score >= 70` → High Performer

`Overall_Score < 70` → Not High Performer

We can represent these classes numerically:

- `1` → High Performer
- `0` → Not High Performer

This creates a binary classification target.

In a real Machine Learning project, the target definition should be based on the actual business or academic problem rather than an arbitrary rule.

`````python
df["High_Performer"] = (
    df["Overall_Score"] >= 70
).astype(int)

df
`````

# Creating the Machine Learning Dataset

We can now separate the dataset into:

`X` → Input Features

`y` → Target Variable

For this example, we will use:

- Age

- Overall Score

## Creating `Family_Size`

The Titanic dataset contains:

- `SibSp` → siblings and spouses aboard
- `Parch` → parents and children aboard

We can combine these values to estimate the total family size.

The passenger themselves should also be counted.

Therefore:

`Family_Size = SibSp + Parch + 1`

This can be a useful feature because family size may have a relationship with survival behavior.

`````python
titanic_features["Family_Size"] = (
    titanic_features["SibSp"] +
    titanic_features["Parch"] +
    1
)

titanic_features[
    ["SibSp", "Parch", "Family_Size"]
].head(10)
`````

## Creating `Fare_Per_Person`

The `Fare` column represents the fare associated with a passenger.

We can calculate the approximate fare per family member:

`Fare_Per_Person = Fare / Family_Size`

This creates a new numerical feature that may provide information different from the original total fare.

`````python
titanic_features["Fare_Per_Person"] = (
    titanic_features["Fare"] /
    titanic_features["Family_Size"]
)

titanic_features[
    ["Fare", "Family_Size", "Fare_Per_Person"]
].head(10)
`````

## Creating `Age_Group`

Age is a numerical variable.

For analysis, it can sometimes be useful to divide age into meaningful groups:

- Child
- Teenager
- Adult
- Senior

We can use `pd.cut()` to divide the numerical age values into intervals.

This creates a categorical variable from a numerical variable.

Note that categorical variables may require additional preprocessing before being passed directly to many Machine Learning algorithms.

`````python
titanic_features["Age_Group"] = pd.cut(
    titanic_features["Age"],
    bins=[0, 12, 18, 60, np.inf],
    labels=[
        "Child",
        "Teenager",
        "Adult",
        "Senior"
    ]
)

titanic_features[
    ["Age", "Age_Group"]
].head(10)
`````

## Renaming a Titanic Column

Real-world datasets often contain column names that may not be descriptive enough for a particular project.

For example, `Pclass` represents passenger class.

We can rename it to `Passenger_Class` to make the meaning clearer.

`````python
titanic_features = titanic_features.rename(
    columns={
        "Pclass": "Passenger_Class"
    }
)

titanic_features.head()
`````

# Common Mistakes

## Mistake 1: Accidentally Overwriting Original Data

Assigning a transformation back to an existing column replaces its previous values.

If the original column is important, create a new column instead.

## Mistake 2: Dropping Important Columns

Always understand why a column exists before removing it.

A column that appears unnecessary may become useful during Feature Engineering or analysis.

## Mistake 3: Creating Meaningless Features

More features do not automatically mean a better Machine Learning model.

Features should have a logical connection to the problem.

## Mistake 4: Data Leakage

A feature should not contain information that would only be available after the prediction event.

For example, if we are predicting whether a student will pass an exam, using the final result of that exam as an input feature would be data leakage.

## Mistake 5: Division by Zero

When creating ratios, always consider whether the denominator can contain zero.

## Mistake 6: Repeatedly Running Transformation Cells

Google Colab allows cells to be executed multiple times.

Repeated transformations can result in unexpected DataFrame changes.

When experimenting with a dataset, keeping an untouched copy can be useful.

## Mistake 7: Inconsistent Column Naming

Use consistent names such as:

- `Study_Hours`
- `Final_Marks`
- `Family_Size`
- `Fare_Per_Person`

Consistent naming makes Data Science workflows easier to maintain.

`````python
import numpy as np
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Age": [20, np.nan, 22, 21, np.nan, 23],
    "Study_Hours": [4, 6, np.nan, 3, 7, 5],
    "Attendance": [75, 82, 91, np.nan, 88, 79],
    "Marks": [62, 74, 89, 55, np.nan, 71]
}

df = pd.DataFrame(data)

df
`````

## 1. Understanding Missing Values

A missing value means that information is not available for a particular observation.

In Pandas, missing numerical values are commonly represented using `NaN`.

For example:

- A student's age may not have been recorded.
- Attendance may be missing.
- A customer's income may not have been provided.

Before deciding how to handle missing values, we should first identify them.

`````python
df.isna()
`````

## 2. Using `isnull()`

`isnull()` performs the same basic task as `isna()`.

It returns:

- `True` → Value is missing
- `False` → Value is present

`isna()` and `isnull()` are equivalent in Pandas.

You can use either one.

`````python
df.isnull()
`````

## 3. Using `notna()`

`notna()` is the opposite of `isna()`.

It returns:

- `True` → Value is present
- `False` → Value is missing

This is useful when we specifically want to identify records containing valid values.

`````python
df.notna()
`````

## 4. Counting Missing Values

A Boolean DataFrame alone is difficult to interpret.

We can count missing values in each column by combining:

- `isna()`
- `sum()`

Because `True` is treated as `1` and `False` as `0`, summing the result gives the number of missing values.

`````python
missing_count = df.isna().sum()

missing_count
`````

## 5. Finding the Total Number of Missing Values

We can calculate the total number of missing cells in the entire DataFrame.

The first `sum()` counts missing values by column.

The second `sum()` adds those column totals together.

`````python
total_missing = df.isna().sum().sum()

print("Total missing values:", total_missing)
`````

## 6. Missing-Value Percentage

The number of missing values alone does not always tell us how serious the problem is.

For example:

- 100 missing values in a dataset of 1,000 rows → 10%
- 100 missing values in a dataset of 100,000 rows → 0.1%

Therefore, we can calculate the percentage of missing values in each column.

This helps us decide whether a column should be:

- Filled
- Removed
- Investigated further

`````python
missing_percentage = (
    df.isna().mean() * 100
)

missing_percentage
`````

## 7. Displaying Missing-Value Information Together

For Data Cleaning, it is useful to see:

- Number of missing values
- Percentage of missing values

together in one table.

`````python
missing_summary = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": df.isna().mean() * 100
})

missing_summary
`````

## 8. Finding Rows with Missing Values

Sometimes we do not want to know only which columns contain missing values.

We want to identify the actual records containing missing information.

We can use `isna().any(axis=1)`.

Here:

`axis=1` means we check across the columns of each row.

A row will be selected if at least one value is missing.

`````python
rows_with_missing = df[
    df.isna().any(axis=1)
]

rows_with_missing
`````

## 9. Finding Rows with No Missing Values

We can also identify complete records.

A complete record contains no missing values in any column.

This can be done using `notna().all(axis=1)`.

`````python
complete_rows = df[
    df.notna().all(axis=1)
]

complete_rows
`````

## 10. Removing Missing Values Using `dropna()`

One approach to missing data is to remove rows containing missing values.

`dropna()` removes rows containing at least one missing value by default.

This method is simple, but it should not be used blindly.

If a dataset contains many missing values, removing rows can result in significant data loss.

`````python
df_without_missing = df.dropna()

df_without_missing
`````

## 11. Dropping Columns with Missing Values

`dropna()` can also be used to remove columns containing missing values.

This can be useful when a column contains an extremely large proportion of missing information.

However, removing a column should only be done after understanding its importance.

`````python
df_without_missing_columns = df.dropna(axis=1)

df_without_missing_columns
`````

## 12. Using `thresh`

Sometimes we do not want to remove every row containing a missing value.

Instead, we can specify the minimum number of non-missing values required.

For example:

`thresh=4`

means that a row must contain at least four non-missing values to remain in the DataFrame.

`````python
df_threshold = df.dropna(
    thresh=5
)

df_threshold
`````

## 13. Filling Missing Values Using `fillna()`

Instead of deleting missing values, we can replace them with another value.

`fillna()` is used for this purpose.

For example, we can replace missing values with `0`.

This is technically easy, but the replacement value should make sense for the variable.

`````python
df_zero_filled = df.fillna(0)

df_zero_filled
`````

## 14. Filling Missing Numerical Values with the Mean

The mean is the arithmetic average.

For a numerical column, we can calculate its mean and use that value to replace missing observations.

Mean imputation can be useful when:

- The data is approximately symmetric
- There are no extreme outliers
- The missing proportion is relatively small

However, mean imputation can be affected by outliers.

`````python
df_mean = df.copy()

mean_age = df_mean["Age"].mean()

df_mean["Age"] = df_mean["Age"].fillna(mean_age)

df_mean
`````

## 15. Filling Missing Values with the Median

The median is the middle value after sorting the data.

Median imputation is often preferred when numerical data contains outliers or is highly skewed.

For example, income data can contain a few extremely high values.

In such situations, the median may represent the typical observation better than the mean.

`````python
df_median = df.copy()

median_age = df_median["Age"].median()

df_median["Age"] = df_median["Age"].fillna(median_age)

df_median
`````

## 16. Mean vs Median

The choice between mean and median depends on the distribution of the data.

### Mean

Useful when the data is reasonably balanced and does not contain severe outliers.

### Median

Often safer when the data is skewed or contains extreme values.

Example:

If salaries are:

- 20,000
- 22,000
- 25,000
- 27,000
- 5,00,000

The extreme value can significantly increase the mean.

The median is less affected by that extreme value.

`````python
salaries = pd.Series([
    20000,
    22000,
    25000,
    27000,
    500000
])

print("Mean:", salaries.mean())
print("Median:", salaries.median())
`````

## 17. Filling Missing Categorical Values with the Mode

The mode is the most frequently occurring value.

Mode imputation is commonly used for categorical columns.

For example, if a column contains:

- Male
- Female
- Female
- Female
- Male
- Missing

we can use the most frequent category to replace the missing value.

The mode can be obtained using `mode()`.

`````python
gender_data = pd.Series([
    "Male",
    "Female",
    "Female",
    np.nan,
    "Female",
    "Male"
])

mode_value = gender_data.mode()[0]

gender_filled = gender_data.fillna(mode_value)

gender_filled
`````

## 18. Forward Fill

Forward fill uses the previous available value to fill a missing value.

The method is:

`ffill`

For example, if the values are:

10

20

Missing

40

the missing value will be replaced with 20.

Forward fill is particularly useful for ordered or time-series data where the previous observation can reasonably represent the missing value.

`````python
values = pd.Series([
    10,
    20,
    np.nan,
    40,
    np.nan,
    60
])

forward_filled = values.ffill()

forward_filled
`````

## 19. Backward Fill

Backward fill uses the next available value to fill a missing value.

The method is:

`bfill`

For example, if the values are:

10

Missing

30

the missing value will be replaced with 30.

Backward fill can be useful when the following observation is more appropriate than the previous observation.

# Common Mistakes

## Mistake 1 — Replacing Everything with Zero

Zero may have a completely different meaning from "missing".

For example:

- Age = 0 is not the same as Age = missing
- Salary = 0 is not necessarily the same as Salary = missing

Do not use zero simply because it is convenient.

## Mistake 2 — Always Using the Mean

Mean imputation is not always the best solution.

If a variable contains significant outliers or is heavily skewed, the median may be more appropriate.

## Mistake 3 — Always Dropping Rows

Removing rows can result in significant data loss.

Always check how many observations will be removed before using `dropna()`.

## Mistake 4 — Filling Categorical Data with the Mean

Mean and median are numerical concepts.

For categorical variables, approaches such as mode imputation may be more appropriate.

## Mistake 5 — Ignoring the Reason for Missingness

Missing data can occur for different reasons.

A missing value may mean:

- Information was not collected
- The value was not applicable
- The user chose not to provide it
- A data collection process failed

Understanding the reason can influence the correct treatment.

## Mistake 6 — Data Leakage During Imputation

In Machine Learning, imputation values such as mean or median should generally be calculated using the training data only.

If the entire dataset is used to calculate the value before train/test splitting, information from the test set can leak into the training process.

This will become important when we learn Machine Learning preprocessing pipelines.

## Mistake 7 — Not Verifying the Result

After handling missing values, always check the dataset again.

Never assume that the cleaning operation worked correctly.

## 1. `sort_values()`

`sort_values()` is used to arrange rows according to the values in one or more columns.

This is useful when we want to:

- Find the highest marks
- Find students with the lowest attendance
- Arrange customers by revenue
- Sort products by price
- Organize records before analysis

The original DataFrame is not changed unless the result is assigned back to it or modified using the appropriate option.

`````python
sorted_df = df.sort_values(
    by="Marks"
)

sorted_df
`````

## 2. Ascending Sorting

Ascending order arranges values from smallest to largest.

For example:

55 → 63 → 74 → 81 → 87 → 92 → 96 → 98

Ascending order is the default behavior of `sort_values()`.

It is useful when looking for:

- Lowest scores
- Lowest prices
- Earliest dates
- Smallest values

`````python
df.sort_values(
    by="Marks",
    ascending=True
)
`````

## 3. Descending Sorting

Descending order arranges values from largest to smallest.

For example:

98 → 96 → 92 → 87 → 81 → 74 → 63 → 55

This is particularly useful when finding:

- Top performers
- Highest sales
- Highest salaries
- Most expensive products
- Largest transactions

`````python
df.sort_values(
    by="Marks",
    ascending=False
)
`````

## 4. Sorting by Multiple Columns

Sometimes one column is not enough to determine the desired order.

We can sort using multiple columns.

For example:

1. Sort by Department
2. Within each department, sort by Marks in descending order

This is useful when comparing records within groups.

`````python
df.sort_values(
    by=["Department", "Marks"],
    ascending=[True, False]
)
`````

## 5. Sorting Data Containing Missing Values

Real-world datasets often contain missing values.

Pandas provides the `na_position` parameter to control where missing values appear after sorting.

Two common options are:

- `"first"` → missing values appear first
- `"last"` → missing values appear last

By default, missing values are placed at the end.

`````python
df_with_missing = df.copy()

df_with_missing.loc[2, "Marks"] = np.nan
df_with_missing.loc[5, "Marks"] = np.nan

print("Missing values at the end:")
print(
    df_with_missing.sort_values(
        by="Marks",
        na_position="last"
    )
)

print("\nMissing values at the beginning:")
print(
    df_with_missing.sort_values(
        by="Marks",
        na_position="first"
    )
)
`````

## 6. `sort_index()`

`sort_index()` sorts the DataFrame according to its index rather than the values of a column.

This is useful when:

- The index order has changed
- We want to restore index ordering
- Working with time-series indexes
- Organizing hierarchical or indexed data

`sort_values()` sorts by column values.

`sort_index()` sorts by index labels.

`````python
shuffled_df = df.sample(
    frac=1,
    random_state=42
)

print("Shuffled DataFrame:")
print(shuffled_df)

print("\nSorted by index:")
print(shuffled_df.sort_index())
`````

`rank()` assigns a numerical position to values.

For example, if students have marks:

96

87

1
2
3
4

Ranking is useful for:

- Student rankings
- Sales rankings
- Product rankings
- Competition results
- Performance analysis

By default, ranking starts from the smallest value.

For performance ranking, we usually want the highest value to receive rank 1, so we can use descending order.

`````python
df["Marks_Rank"] = df["Marks"].rank(
    ascending=False
)

df
`````

## 8. Handling Ties in Ranking

Two or more records can have the same value.

Pandas provides different ranking methods for handling ties.

Common methods include:

- `average`
- `min`
- `max`
- `first`
- `dense`

The default method is `average`.

The choice depends on how the ranking should behave when multiple observations have the same value.

`````python
rank_example = pd.Series([
    95,
    90,
    90,
    85,
    80
])

print("Average ranking:")
print(
    rank_example.rank(
        ascending=False
    )
)

print("\nDense ranking:")
print(
    rank_example.rank(
        ascending=False,
        method="dense"
    )
)
`````

## 9. Finding Top and Bottom Records

Pandas provides convenient methods for finding extreme observations.

`nlargest()` returns the rows with the largest values.

`nsmallest()` returns the rows with the smallest values.

These methods are useful when we want:

- Top 5 students
- Bottom 10 products
- Highest revenue transactions
- Lowest-performing records

They can be more convenient than manually sorting the entire DataFrame.

`````python
top_3_students = df.nlargest(
    3,
    "Marks"
)

bottom_3_students = df.nsmallest(
    3,
    "Marks"
)

print("Top 3 Students:")
print(top_3_students)

print("\nBottom 3 Students:")
print(bottom_3_students)
`````

## 10. Understanding Duplicate Data

A duplicate record occurs when the same information appears more than once.

Duplicates can occur because of:

- Repeated data entry
- Multiple file imports
- Database synchronization problems
- Copy-paste errors
- Combining datasets
- Data collection errors

Duplicate records can distort:

- Counts
- Averages
- Statistics
- Machine Learning models

Therefore, duplicate detection is an important Data Cleaning step.

`````python
duplicate_data = {
    "Student_ID": [101, 102, 103, 103, 104, 105, 105],
    "Name": [
        "A", "B", "C", "C",
        "D", "E", "E"
    ],
    "Marks": [
        78, 85, 91, 91,
        67, 88, 88
    ]
}

duplicates_df = pd.DataFrame(
    duplicate_data
)

duplicates_df
`````

## 11. Detecting Duplicate Rows with `duplicated()`

`duplicated()` identifies duplicate rows.

It returns:

- `False` → First occurrence
- `True` → Duplicate occurrence

By default, Pandas considers the first occurrence as the original and subsequent identical rows as duplicates.

`````python
duplicates_df.duplicated()
`````

## 12. Displaying Duplicate Rows

We can use the Boolean result from `duplicated()` to display only duplicate records.

This allows us to inspect the duplicate data before deciding whether it should be removed.

`````python
duplicate_rows = duplicates_df[
    duplicates_df.duplicated()
]

duplicate_rows
`````

## 13. Detecting Duplicates Using Selected Columns

Sometimes two rows may not be completely identical but may represent the same entity.

For example, if `Student_ID` is supposed to uniquely identify a student, we can check duplicates using only that column.

The `subset` parameter allows us to specify which columns should be considered when identifying duplicates.

`````python
duplicates_df[
    duplicates_df.duplicated(
        subset=["Student_ID"]
    )
]
`````

## 14. Removing Duplicate Rows

`drop_duplicates()` removes duplicate records.

By default, it keeps the first occurrence and removes later duplicates.

Before removing duplicates from a real dataset, it is good practice to inspect them first.

`````python
clean_duplicates = duplicates_df.drop_duplicates()

clean_duplicates
`````

## 15. Keeping the Last Duplicate

Sometimes the most recent occurrence should be preserved instead of the first.

`keep="last"` keeps the final occurrence and removes earlier duplicates.

This can be useful when later records contain updated information.

`````python
keep_last = duplicates_df.drop_duplicates(
    subset=["Student_ID"],
    keep="last"
)

keep_last
`````

## 16. Removing All Records That Have Duplicates

Sometimes we do not want to keep either occurrence.

Using `keep=False` marks every record belonging to a duplicated group for removal.

This is useful when duplicate records are ambiguous and none should be treated as the correct record automatically.

`````python
unique_only = duplicates_df.drop_duplicates(
    subset=["Student_ID"],
    keep=False
)

unique_only
`````

## 1. Understanding `groupby()`

`groupby()` divides a DataFrame into groups based on the values of one or more columns.

For example, our dataset contains students from:

- AIML
- CSE

We can group all students according to their department.

The important idea is:

```text
DataFrame
    ↓
Group by Department
    ↓
AIML Group
CSE Group
    ↓
Perform Calculation

groupby() is generally used together with aggregation functions such as:

mean()
sum()
min()
max()
count()

`````python


department_groups = df.groupby("Department")

department_groups
`````

## 2. Grouping by One Column and Calculating the Mean

One of the most common uses of `groupby()` is calculating averages for each group.

For example, we can calculate the average marks for each department.

This allows us to compare groups instead of looking at individual students.

`````python
average_marks = df.groupby("Department")["Marks"].mean()

average_marks
`````

## 3. Grouping by Multiple Columns

We can group data using more than one column.

For example, we can group students by:

- Department
- Year

This creates groups such as:

- AIML + First
- AIML + Second
- AIML + Third
- CSE + First
- CSE + Second
- CSE + Third

`````python
department_year_marks = (
    df
    .groupby(["Department", "Year"])["Marks"]
    .mean()
)

department_year_marks
`````

## 4. `sum()`

`sum()` calculates the total value within each group.

For example, we can calculate the total study hours for each department.

This is useful for measuring totals such as:

- Total sales
- Total revenue
- Total study hours
- Total units sold
- Total expenses

`````python
total_study_hours = (
    df
    .groupby("Department")["Study_Hours"]
    .sum()
)

total_study_hours
`````

## 5. `mean()`

`mean()` calculates the arithmetic average within each group.

For example, we can calculate average attendance by department.

Mean is commonly used when comparing the typical value across groups.

`````python
average_attendance = (
    df
    .groupby("Department")["Attendance"]
    .mean()
)

average_attendance
`````

## 6. `median()`

`median()` returns the middle value of each group after sorting the observations.

Median is useful when the data contains outliers or is heavily skewed.

For example, salary data often contains extreme values, so median salary can sometimes be more representative than mean salary.

`````python
median_marks = (
    df
    .groupby("Department")["Marks"]
    .median()
)

median_marks
`````

## 7. `min()` and `max()`

`min()` returns the smallest value in each group.

`max()` returns the largest value in each group.

These operations are useful when we want to understand the range of values within each category.

`````python
minimum_marks = (
    df
    .groupby("Department")["Marks"]
    .min()
)

maximum_marks = (
    df
    .groupby("Department")["Marks"]
    .max()
)

print("Minimum Marks:")
print(minimum_marks)

print("\nMaximum Marks:")
print(maximum_marks)
`````

`count()` counts the number of non-missing values in each group.

This is important because `count()` does not simply count rows.

If a column contains missing values, those missing values are excluded from the count.

This makes `count()` particularly useful when analyzing data completeness.

`````python
marks_count = (
    df
    .groupby("Department")["Marks"]
    .count()
)

marks_count
`````

## 9. `size()`

`size()` counts the total number of rows in each group.

Unlike `count()`, it includes rows even when some values are missing.

The difference is important:

`count()` → Counts non-missing values

`size()` → Counts rows

`````python
group_size = (
    df
    .groupby("Department")
    .size()
)

group_size
`````

## 10. Understanding `count()` vs `size()`

Consider a dataset where one student has a missing marks value.

`size()` will still count the student because the row exists.

`````python
df_with_missing = df.copy()

df_with_missing.loc[2, "Marks"] = np.nan

print("count():")
print(
    df_with_missing
    .groupby("Department")["Marks"]
    .count()
)

print("\nsize():")
print(
    df_with_missing
    .groupby("Department")
    .size()
)
`````

# Common Mistakes

## Mistake 1 — Confusing `count()` and `size()`

`count()` counts non-missing values.

`size()` counts rows.

They may produce different results when missing values exist.

## Mistake 2 — Forgetting `reset_index()`

After grouping, the grouping column may become the index.

Use `reset_index()` when you need a regular DataFrame structure.

## Mistake 3 — Calculating the Wrong Average

Always understand what the average represents.

An average calculated over the wrong grouping can produce misleading conclusions.

## Mistake 4 — Using `mean()` on Inappropriate Columns

Not every column has a meaningful numerical average.

For example, taking the average of a person's name has no meaningful interpretation.

## Mistake 5 — Ignoring Group Size

A group with an average based on 2 observations should not necessarily be interpreted with the same confidence as a group containing 10,000 observations.

Always consider the number of observations in each group.

## Mistake 6 — Group-Level Data Leakage

When using grouped statistics as Machine Learning features, make sure information from validation or test data does not leak into the training process.

## Mistake 7 — Assuming Correlation or Causation

A difference between groups does not automatically mean one variable caused the difference.

GroupBy helps us discover patterns; further statistical analysis is required to establish stronger conclusions.

`````python
import numpy as np
import pandas as pd

students_1 = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["A", "B", "C"],
    "Marks": [78, 85, 92]
})

students_2 = pd.DataFrame({
    "Student_ID": [104, 105, 106],
    "Name": ["D", "E", "F"],
    "Marks": [67, 88, 74]
})

print("Students 1:")
print(students_1)

print("\nStudents 2:")
print(students_2)
`````

## 1. `concat()`

`concat()` is used to concatenate multiple Pandas objects.

It is commonly used when datasets have the same structure and need to be combined.

For example, suppose:

- `students_1` contains students 101–103
- `students_2` contains students 104–106

Both DataFrames contain the same columns.

We can combine them row by row.

This is called **row-wise concatenation**.

`````python
combined_students = pd.concat(
    [students_1, students_2]
)

combined_students
`````

## 2. Resetting the Index After Concatenation

When DataFrames are concatenated, Pandas preserves the original indexes.

This can result in repeated index values.

If we want a fresh sequential index, we can use:

`ignore_index=True`

This is commonly useful when combining datasets from multiple files.

`````python
combined_students = pd.concat(
    [students_1, students_2],
    ignore_index=True
)

combined_students
`````

## 3. Column-Wise Concatenation

`concat()` can also combine DataFrames horizontally.

This means the columns are placed next to each other.

This is useful when two DataFrames:

- Have the same number of rows
- Represent different features
- Need to be combined side by side

For example, one DataFrame may contain student information while another contains academic information.

`````python
student_info = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["A", "B", "C"]
})

academic_info = pd.DataFrame({
    "Study_Hours": [5, 7, 8],
    "Attendance": [80, 90, 85]
})

combined_columns = pd.concat(
    [student_info, academic_info],
    axis=1
)

combined_columns
`````

## 4. Understanding `axis`

The `axis` parameter determines the direction of concatenation.

`axis=0`

Means rows are combined vertically.

`axis=1`

Means columns are combined horizontally.

Therefore:

`axis=0` → Add more rows

`axis=1` → Add more columns

`````python
vertical = pd.concat(
    [students_1, students_2],
    axis=0,
    ignore_index=True
)

horizontal = pd.concat(
    [student_info, academic_info],
    axis=1
)

print("Vertical:")
print(vertical)

print("\nHorizontal:")
print(horizontal)
`````

## 5. Concatenating DataFrames with Different Columns

Real-world datasets may not always contain exactly the same columns.

When DataFrames with different columns are concatenated, Pandas creates the union of all columns.

Values that do not exist in a particular DataFrame become missing values.

This is important because combining datasets can introduce new missing values.

`````python
batch_1 = pd.DataFrame({
    "Student_ID": [101, 102],
    "Name": ["A", "B"],
    "Marks": [80, 90]
})

batch_2 = pd.DataFrame({
    "Student_ID": [103, 104],
    "Name": ["C", "D"],
    "Attendance": [85, 92]
})

different_columns = pd.concat(
    [batch_1, batch_2],
    ignore_index=True
)

different_columns
`````

# `merge()`

`merge()` is used when two DataFrames contain related information that should be combined using one or more common columns.

For example:

Students:

`Student_ID | Name`

Marks:

`Student_ID | Marks`

Both tables contain `Student_ID`.

We can use that common column to combine the information.

This is similar to SQL joins.

`````python
student_details = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["A", "B", "C", "D"],
    "Department": ["AIML", "CSE", "AIML", "ECE"]
})

student_marks = pd.DataFrame({
    "Student_ID": [101, 102, 103, 105],
    "Marks": [78, 85, 92, 88]
})

print("Student Details:")
print(student_details)

print("\nStudent Marks:")
print(student_marks)
`````

## 6. Inner Join

An inner join keeps only records that have matching keys in both DataFrames.

In this example:

Student IDs in the first DataFrame:

101, 102, 103, 104

Student IDs in the second DataFrame:

101, 102, 103, 105

The common IDs are:

101, 102, 103

Therefore, an inner join keeps only these three students.

Inner joins are useful when we only want records that exist in both datasets.

`````python
inner_merged = pd.merge(
    student_details,
    student_marks,
    on="Student_ID",
    how="inner"
)

inner_merged
`````

## 7. Left Join

A left join keeps every record from the left DataFrame.

Matching information from the right DataFrame is added when available.

If there is no matching record, Pandas inserts a missing value.

In this example, Student ID 104 exists in the student details table but not in the marks table.

Therefore, Student 104 remains in the result but has a missing Marks value.

`````python
left_merged = pd.merge(
    student_details,
    student_marks,
    on="Student_ID",
    how="left"
)

left_merged
`````

## 8. Right Join

A right join keeps every record from the right DataFrame.

Matching information from the left DataFrame is added when available.

In this example, Student ID 105 exists in the marks table but not in the student details table.

Therefore, Student 105 remains in the result with missing information from the left DataFrame.

`````python
right_merged = pd.merge(
    student_details,
    student_marks,
    on="Student_ID",
    how="right"
)

right_merged
`````

## 9. Outer Join

An outer join keeps all records from both DataFrames.

If a matching record does not exist, Pandas inserts a missing value.

In this example:

- Student 104 exists only in the first table.
- Student 105 exists only in the second table.

Both students will remain in the final result.

Outer joins are useful when we want to preserve all information from both datasets.

`````python
outer_merged = pd.merge(
    student_details,
    student_marks,
    on="Student_ID",
    how="outer"
)

outer_merged
`````

## 10. Understanding Join Types

The four major join types are:

### Inner

Keep matching records only.

### Left

Keep everything from the left DataFrame.

### Right

Keep everything from the right DataFrame.

### Outer

Keep everything from both DataFrames.

A useful way to remember them:

`inner` → Intersection

`left` → Everything on left

`right` → Everything on right

`outer` → Union

`````python
join_types = {
    "inner": pd.merge(
        student_details,
        student_marks,
        on="Student_ID",
        how="inner"
    ),
    "left": pd.merge(
        student_details,
        student_marks,
        on="Student_ID",
        how="left"
    ),
    "right": pd.merge(
        student_details,
        student_marks,
        on="Student_ID",
        how="right"
    ),
    "outer": pd.merge(
        student_details,
        student_marks,
        on="Student_ID",
        how="outer"
    )
}

for name, result in join_types.items():
    print(f"\n{name.upper()} JOIN")
    print(result)
`````

## 11. Merging Using Different Column Names

Sometimes the key columns represent the same information but have different names.

For example:

First DataFrame:

`Student_ID`

Second DataFrame:

`ID`

We can specify:

- `left_on`
- `right_on`

to tell Pandas which columns should be matched.

`````python
students = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["A", "B", "C"]
})

marks = pd.DataFrame({
    "ID": [101, 102, 103],
    "Marks": [78, 85, 92]
})

merged_different_names = pd.merge(
    students,
    marks,
    left_on="Student_ID",
    right_on="ID",
    how="inner"
)

merged_different_names
`````

Sometimes a single column is not enough to uniquely identify a record.

# Common Mistakes

## Mistake 1 — Using `concat()` When a `merge()` Is Required

Use `concat()` when datasets need to be stacked.

Use `merge()` when datasets need to be connected using a key.

## Mistake 2 — Choosing the Wrong Join Type

An inner join can remove records that do not have a match.

A left join preserves all records from the left DataFrame.

Always understand which records you need to preserve.

## Mistake 3 — Joining on the Wrong Column

The join key must represent the same entity in both datasets.

For example, joining customers using their names can be dangerous because names may not be unique.

A unique identifier such as Customer ID is usually safer.

## Mistake 4 — Unexpected Row Multiplication

If the join key appears multiple times in both DataFrames, a merge can create multiple combinations.

This can unexpectedly increase the number of rows.

Always check the uniqueness of your join keys.

## Mistake 5 — Ignoring Missing Values After a Join

Left, right, and outer joins can introduce missing values when matching records do not exist.

The resulting DataFrame should be checked after merging.

## Mistake 6 — Duplicate Columns

If both DataFrames contain columns with the same name, Pandas may add suffixes such as `_x` and `_y`.

These columns should be reviewed carefully.

## Mistake 7 — Data Leakage

When combining datasets for Machine Learning, do not accidentally merge information that would not have been available at prediction time.

## Mistake 8 — Not Checking the Result

After combining datasets, always inspect:

- Number of rows
- Number of columns
- Missing values
- Duplicate records
- Join keys
- Unexpected values

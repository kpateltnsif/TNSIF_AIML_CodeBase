# What is a Dictionary?

A **dictionary** is a collection of **key-value pairs**.

Each key is unique and is associated with a value.

General form:

```text
Key  →  Value
```

Examples:

```text
Name      → Rahul
Age       → 21
City      → Nashik
CGPA      → 8.75
```

Unlike lists and tuples, dictionaries access data using **keys** instead of indices.

# Characteristics of Dictionaries

- Store data as **key-value pairs**
- Keys must be unique
- Values can be duplicated
- Mutable (can be modified)
- Ordered (Python 3.7+)
- Fast lookup using keys
- Do not support integer indexing like lists

# List vs Tuple vs Set vs Dictionary

| Feature | List | Tuple | Set | Dictionary |
|---------|------|-------|-----|------------|
| Ordered | ✅ | ✅ | ❌ | ✅ |
| Mutable | ✅ | ❌ | ✅ | ✅ |
| Duplicates | ✅ | ✅ | ❌ | Keys ❌ / Values ✅ |
| Indexing | ✅ | ✅ | ❌ | By Key |
| Uses | Collections | Fixed Data | Unique Data | Key-Value Data |

# Accessing Dictionary Values

Unlike lists, dictionaries do **not** use numeric indexing.

Instead, values are accessed using their **keys**.

### Syntax

```python
dictionary[key]
```

or

```python
dictionary.get(key)
```

Both methods retrieve values, but they behave differently when a key does not exist.

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

print(student["Name"])
print(student["Age"])
print(student["City"])
````

Each key returns its corresponding value.

```python
student["Name"]
```

returns

```text
Rahul
```

````python
model = {
    "Algorithm": "Random Forest",
    "Accuracy": 95.4,
    "Features": 18
}

print("Algorithm :", model["Algorithm"])
print("Accuracy :", model["Accuracy"])
````

# get()

The `get()` method safely retrieves values.

### Syntax

```python
dictionary.get(key)
```

If the key does not exist,

`get()` returns **None** instead of raising an error.

````python
student = {
    "Name": "Rahul",
    "Age": 21
}

print(student.get("Name"))

print(student.get("Age"))
````

````python
student = {
    "Name": "Rahul"
}

print(student.get("City"))
````

You can also specify a default value.

```python
student.get("City", "Not Available")
```

````python
student = {
    "Name": "Rahul"
}

print(student.get("City", "Not Available in the Dictionary "))
````

# update()

The `update()` method adds or updates multiple key-value pairs.

### Syntax

```python
dictionary.update(other_dictionary)
```

````python
student = {
    "Name": "Rahul"
}

student.update({
    "Age": 21,
    "City": "Nashik"
})

print(student)
````

````python
student = {
    "Name": "Rahul",
    "Age": 20
}

student.update({
    "Age": 21
})

print(student)
````

# pop()

Removes a key and returns its value.

### Syntax

```python
dictionary.pop(key)
```

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

removed = student.pop("City")

print("Removed :", removed)
print(student)
````

# popitem()

Removes the **last inserted** key-value pair.

Returns both the key and value as a tuple.

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

item = student.popitem()

print(item)

print(student)
````

# del Statement

Use `del` to delete a specific key or the entire dictionary.

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

del student["Age"]

print(student)
````

````python
student = {
    "Name": "Rahul"
}

del student

# Uncomment to verify

print(student)
````

# clear()

Removes all key-value pairs.

The dictionary still exists.

````python
student = {
    "Name": "Rahul",
    "Age": 21
}

student.clear()

print(student)
````

````python
metrics = {
    "Accuracy": 91.5,
    "Precision": 90.8
}

metrics["Recall"] = 92.4

metrics["Accuracy"] = 94.3

print(metrics)
````

# keys()

The `keys()` method returns all keys in the dictionary.

### Syntax

```python
dictionary.keys()
```

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

print(student.keys())
````

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

print(list(student.keys()))
````

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

print(student.values())
````

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

print(list(student.values()))
````

# items()

The `items()` method returns each key-value pair as a tuple.

### Syntax

```python
dictionary.items()
```

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

print(student.items())
````

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

print(list(student.items()))
````

# Traversing Dictionaries

There are multiple ways to iterate through a dictionary.

Each method is useful depending on whether you need:

- Keys
- Values
- Both keys and values

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

for key in student:
    print(key)
````

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

for key in student.keys():
    print(key)
````

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

for value in student.values():
    print(value)
````

# Best Practice

When you need both keys and values,

use `items()`.

````python
student = {
    "Name": "Rahul",
    "Age": 21,
    "City": "Nashik"
}

for key, value in student.items():
    print(key, ":", value)
````

# Nested Dictionaries

A dictionary can contain other dictionaries.

This is called a **Nested Dictionary**.

Nested dictionaries are commonly used to represent structured data.

````python
students = {
    "Student1": {
        "Name": "Rahul",
        "CGPA": 8.9
    },
    "Student2": {
        "Name": "Aisha",
        "CGPA": 9.3
    }
}

print(students)
````

````python
students = {
    "Student1": {
        "Name": "Rahul",
        "CGPA": 8.9
    },
    "Student2": {
        "Name": "Aisha",
        "CGPA": 9.3
    }
}

print(students["Student1"]["Name"])

print(students["Student2"]["CGPA"])
````

# Dictionary Comprehension

Dictionary comprehension provides a concise way to create dictionaries.

### Syntax

```python
{
    key: value
    for item in iterable
}
```

It works similarly to list comprehension.

````python
squares = {
    x: x ** 2
    for x in range(1, 6)
}

print(squares)
````

````python
even = {
    x: x**2
    for x in range(10)
    if x % 2 == 0
}

print(even)
````

````python
students = [
    "Rahul",
    "Aisha",
    "John"
]

marks = [
    91,
    88,
    95
]

student_marks = dict(zip(students, marks))

print(student_marks)
````

#Practical AI & Data Science Examples

````python
model = {
    "Algorithm": "Random Forest",
    "Trees": 100,
    "Max Depth": 12,
    "Learning Rate": 0.01
}

for parameter, value in model.items():
    print(parameter, ":", value)
````

````python
dataset = {
    "Rows": 5000,
    "Columns": 18,
    "Target": "Purchased",
    "Missing Values": 42
}

for key, value in dataset.items():
    print(key, ":", value)
````

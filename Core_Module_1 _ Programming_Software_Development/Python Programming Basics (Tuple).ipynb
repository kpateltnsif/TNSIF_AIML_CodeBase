# What is a Tuple?

A **tuple** is an ordered collection of elements.

Like a list, a tuple can store:

- Integers
- Floats
- Strings
- Boolean values
- Other tuples
- Mixed data types

However, unlike a list,

**tuples are immutable.**

Once created, their elements cannot be modified.

# List vs Tuple

| List | Tuple |
|------|--------|
| Mutable | Immutable |
| Uses `[]` | Uses `()` |
| Slower | Faster |
| Larger memory | Less memory |
| Used for changing data | Used for fixed data |

Use tuples whenever the data should not change.

# Creating Tuples

Tuples are created using parentheses.

### Syntax

```python
tuple_name = (item1, item2, item3)
```

````python
numbers = (10, 20, 30, 40)

print(numbers)
````

````python
courses = (
    "Python",
    "Machine Learning",
    "Deep Learning"
)

print(courses)
````

````python
mixed = (
    10,
    "Python",
    3.14,
    True
)

print(mixed)
````

# Single Element Tuple

This is a common interview question.

```python
number = (10)
```

Is **not** a tuple.

It is simply an integer.

To create a tuple with one element, include a trailing comma.

````python
a = (10)

b = (10,)

print(type(a))

print(type(b))
````

# Length of a Tuple

Use the built-in `len()` function.

````python
features = (
    "Age",
    "Salary",
    "Experience",
    "Education"
)

print(len(features))
````

# Checking the Data Type

````python
numbers = (10,20,30)

print(type(numbers))
````

# Indexing

Tuple indexing works exactly like lists.

```text
Element : 10   20   30   40

Index   : 0    1    2    3
```

````python
numbers = (10,20,30,40)

print(numbers[0])

print(numbers[2])

print(numbers[-1])
````

````python
numbers = (10,20,30,40,50,60)

print(numbers[:3])

print(numbers[2:])

print(numbers[::-1])
````

Tuple slicing works exactly like list slicing.

The result of slicing is another tuple.

# Tuples are Immutable

Once created,

their elements cannot be changed.

````python
numbers = (10,20,30)

numbers[1] = 100
````

Output

```text
TypeError
```

This happens because tuples do not support item assignment.

````python
feature_names = (
    "Age",
    "Salary",
    "Experience",
    "Purchased"
)

print(feature_names)
````

Feature names rarely change during model training.

Therefore,

a tuple is a good choice for storing them.

✔ Tuples are ordered.

# Tuple Operations

Although tuples are immutable, they support many useful operations.

You can:

- Concatenate tuples
- Repeat tuples
- Check membership
- Traverse tuples
- Search elements
- Count elements
- Find indices
- Perform packing and unpacking

These operations make tuples useful for storing fixed collections of data.

# Tuple Concatenation

Use the `+` operator to combine two tuples.

A new tuple is created because tuples are immutable.

````python
tuple1 = (10, 20, 30)

tuple2 = (40, 50)

result = tuple1 + tuple2

print(result)
````

# Tuple Repetition

Use the `*` operator to repeat a tuple.

````python
numbers = (1, 2, 3)

print(numbers * 3)
````

# Membership Operators

Use:

- `in`
- `not in`

to check whether an element exists inside a tuple.

````python
courses = (
    "Python",
    "Machine Learning",
    "Deep Learning"
)

print("Python" in courses)

print("Java" in courses)

print("Java" not in courses)
````

# Tuple Methods

Unlike lists, tuples have only **two built-in methods**.

| Method | Purpose |
|---------|----------|
| `count()` | Counts occurrences |
| `index()` | Returns first matching index |

````python
labels = (1, 0, 1, 1, 0, 1)

print(labels.count(1))

print(labels.count(0))
````

> ## 🧠 AI Example
>
> A tuple of class labels can be analyzed using `count()` to determine the number of samples belonging to each class.

````python
courses = (
    "Python",
    "Machine Learning",
    "Deep Learning"
)

print(courses.index("Machine Learning"))
````

If the value does not exist,

Python raises a `ValueError`.

# Traversing a Tuple

Like lists, tuples can be traversed using loops.

````python
numbers = (10, 20, 30, 40)

for number in numbers:
    print(number)
````

````python
courses = (
    "Python",
    "Machine Learning",
    "Deep Learning"
)

for index, course in enumerate(courses):
    print(index, course)
````

# Tuple Packing

Packing means storing multiple values into a single tuple.

Example:

```python
student = ("Rahul", 20, "AIML")
```

Python automatically packs these values into one tuple.

````python
student = "Rahul", 20, "AIML", 8.95

print(student)

print(type(student))
````

# Tuple Unpacking

Unpacking means assigning each element of a tuple to separate variables.

### Syntax

```python
a, b, c = tuple_name
```

````python
student = (
    "Rahul",
    20,
    "AIML"
)

name, age, branch = student

print(name)

print(age)

print(branch)
````

The number of variables must match the number of elements.

Otherwise, Python raises a `ValueError`.

````python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

print(first)

print(middle)

print(last)
````

The `*` operator collects the remaining elements into a list.

This is called **Extended Iterable Unpacking**.

# Nested Tuples

Tuples can contain other tuples.

This is useful when storing structured data.

````python
dataset = (
    (21, 35000, 0),
    (25, 52000, 1),
    (30, 65000, 1)
)

print(dataset)
````

````python
dataset = (
    (21, 35000, 0),
    (25, 52000, 1),
    (30, 65000, 1)
)

print(dataset[0])

print(dataset[1][1])

print(dataset[2][2])
````

Nested tuples can represent fixed datasets where records are not expected to change.

# Built-in Functions

Most built-in functions that work on lists also work on tuples.

````python
numbers = (45, 12, 89, 34, 67)

print("Length :", len(numbers))

print("Maximum :", max(numbers))

print("Minimum :", min(numbers))

print("Sum :", sum(numbers))

print("Sorted :", sorted(numbers))
````

Notice that:

```python
sorted()
```

returns a **list**, not a tuple.

````python
accuracy = (
    72.4,
    81.6,
    89.2,
    93.5,
    95.1
)

print("Best Accuracy :", max(accuracy))

print("Average Accuracy :", sum(accuracy) / len(accuracy))
````

````python
feature_names = (
    "Age",
    "Salary",
    "Experience",
    "Education"
)

for feature in feature_names:
    print(feature)
````

````python
labels = (1, 0, 1, 1, 0, 0, 1)

print("Class 1 :", labels.count(1))

print("Class 0 :", labels.count(0))
````

# Interview Tips

### Q1. Why are tuples faster than lists?

Because tuples are immutable, Python can optimize their storage and access.

---

### Q2. When should you use a tuple?

Use tuples when the data should remain constant throughout the program.

---

### Q3. Can a tuple contain a list?

Yes.

Although the tuple itself cannot be modified, it can contain mutable objects like lists.

Example:

```python
data = (10, [20, 30], "Python")
```

The list inside the tuple can still be modified.

# Quick Recap

✔ Tuples support concatenation and repetition.

✔ Membership operators (`in`, `not in`) work with tuples.

✔ Tuples have only two methods: `count()` and `index()`.

✔ Packing groups multiple values into a tuple.

✔ Unpacking assigns tuple elements to variables.

✔ Extended unpacking uses `*` to capture remaining values.

✔ Nested tuples are useful for fixed datasets.

✔ Most built-in functions (`len()`, `max()`, `min()`, `sum()`) work with tuples.

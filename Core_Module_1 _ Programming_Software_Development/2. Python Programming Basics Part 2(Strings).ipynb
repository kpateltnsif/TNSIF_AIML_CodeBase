# String Indexing

Each character in a string has a unique position called an **index**.

Python uses **zero-based indexing**, meaning counting starts from **0**.

Consider the string:

```text
P  Y  T  H  O  N
```

Positive Index:

```text
Character :  P   Y   T   H   O   N
Index     :  0   1   2   3   4   5
```

Every character can be accessed using its index.

````python
language = "Python"

print(language[0])
print(language[1])
print(language[2])
print(language[5])
````

# Negative Indexing

Python also supports negative indexing.

Negative indexing starts from the **end** of the string.

```text
Character :  P   Y   T   H   O   N
Index     : -6  -5  -4  -3  -2  -1
```

`-1` always refers to the last character.

````python
language = "Python"

print(language[-1])
print(language[-2])
print(language[-3])
````

# What Happens if the Index Doesn't Exist?

If you try to access an index outside the string's range, Python raises an error.

Example:

```python

```

Output:

```text

```

Always ensure the index is within the valid range.

# String Slicing

## What is String Slicing?

Sometimes we don't want the entire string.

Instead, we only need a portion of it.

For example:

- First Name from Full Name
- Domain from an Email Address
- First 5 characters of a Product ID
- First word of a sentence

Python provides **String Slicing** to extract a part of a string.

Instead of accessing one character at a time, slicing allows us to retrieve multiple characters together.

# Syntax of String Slicing

```python
string[start : stop]
```

Where:

- **start** → Starting index (included)
- **stop** → Ending index (excluded)

> **Important:** The `stop` index is **not included** in the result.

For example,

```python
text = "Python"

text[0:4]
```

returns:

```text
Pyth
```

The character at index **4** is **not** included.

# Visual Representation

Consider the following string:

```text
Character :  P   y   t   h   o   n
Index     :  0   1   2   3   4   5
```

Now observe:

```python
text[1:4]
```

```text
Character :      y   t   h
Index     :      1   2   3
```

Result:

```text
yth
```

````python
language = "Python"

print(language[0:2])
print(language[0:4])
print(language[2:5])
print(language[1:6])
````

### Expected Output

```text
Py
Pyth
tho
ython
```

# Default Values in Slicing

Python allows you to omit the `start` or `stop` index.

If omitted:

- `start` defaults to **0**
- `stop` defaults to the **length of the string**

````python
language = "Python"

print(language[:3])
print(language[2:])
print(language[:])
````

### Expected Output

```text
Pyt
thon
Python
```

Explanation:

- `[:3]` → From the beginning up to index 3 (exclusive)
- `[2:]` → From index 2 to the end
- `[:]` → Entire string

# Negative Index Slicing

Negative indices can also be used while slicing.

Recall:

```text
Character :  P   y   t   h   o   n
Index     : -6 -5 -4 -3 -2 -1
```

Negative slicing starts counting from the end of the string.

````python
language = "Python"

print(language[-4:])
print(language[:-2])
print(language[-5:-2])
````

### Expected Output

```text
thon
Pyth
yth
```

# Slice Step

Slicing also supports a third parameter called **step**.

Syntax:

```python
string[start : stop : step]
```

The **step** tells Python how many characters to skip while slicing.

By default:

```python
step = 1
```

````python
language = "PythonProgramming"

print(language[0::2])
print(language[::3])
````

### Expected Output

```text
PtoPormig
PhPg
```

Python selects every 2nd or every 3rd character depending on the step value.

# Reversing a String

One of the most useful applications of slicing is reversing a string.

Use a step value of **-1**.

````python
language = "Python"

print(language[::-1])
````

### Expected Output

```text
nohtyP
```

The negative step moves through the string from right to left.

**Concatenation** means joining two or more strings together.

Python uses the `+` operator for concatenation.

### Syntax

```python

```

The result is a new string containing both strings joined together.

````python
first_name = "Rahul "
last_name = "Sharma"

full_name = first_name + last_name

print(full_name)
````

# String Immutability

## Learning Objectives

After completing this section, you will be able to:

- Understand what immutability means.
- Explain why strings are immutable.
- Identify operations that create new strings.
- Differentiate between modifying a string and creating a new one.
- Understand how Python manages string objects in memory.

---

One of the most important properties of Python strings is that they are **immutable**.

**Immutable** means:

> **Once a string is created, its contents cannot be changed.**

You can create a new string, but you cannot modify an existing string.

# What Does "Immutable" Mean?

Imagine writing a sentence using a permanent marker on a whiteboard.

If you want to change one word, you cannot erase only that word.

Instead, you write the entire sentence again.

Python strings work in a similar way.

Whenever you "change" a string, Python actually creates a **new string object**.

# Memory Representation

Suppose we write:

```python
language = "Python"
```

Internally, Python stores:

```text
language
    │
    ▼
+------------+
|  "Python"  |
+------------+
```

The variable `language` refers to the string object stored in memory.

# Can We Modify a Character?

Suppose we want to change:

```text
Python
```

to

```text
Jython
```

A beginner might try:

```python
language = "Python"

language[0] = "J"
```

Let's see what happens.

````python
language = "Python"

language[0] = "J"
````

### Output

```text
TypeError: 'str' object does not support item assignment
```

Python raises an error because individual characters of a string **cannot be modified**.

# Why Does This Error Occur?

Remember:

```python
language[0]
```

refers to the first character.

Python allows us to **read** characters using indexing, but **not modify** them.

Therefore,

```python
language[0] = "J"
```

is not allowed.

# The Correct Way

Instead of modifying the existing string, create a new one.

Example:

```python
language = "Python"

language = "Jython"
```

Now the variable points to a completely new string object.

````python
language = "Python"

print(language)

language = "Jython"

print(language)
````

# Memory After Reassignment

Initially:

```text
language
    │
    ▼
+------------+
|  "Python"  |
+------------+
```

After:

```python
language = "Jython"
```

```text
language
    │
    ▼
+------------+
|  "Jython"  |
+------------+
```

The original `"Python"` object remains unchanged.

If nothing references it anymore, Python's garbage collector may remove it.

# Why Does `replace()` Work?

Earlier, we used:

```python
text = "I love Java"

new_text = text.replace("Java", "Python")
```

Did `replace()` modify the original string?

**No.**

It created a **new string**.

````python
text = "I love Java"

new_text = text.replace("Java", "Python")

print(text)
print(new_text)
````

### Output

```text
I love Java
I love Python
```

Notice that the original string remains unchanged.

# Concatenation and Immutability

Consider the following code:

```python
name = "AI"

name = name + " ML"
```

Did Python modify the original string?

No.

It created a brand-new string containing:

```text
AI ML
```

````python
name = "AI"

print(id(name))

name = name + " ML"

print(id(name))
````

The two memory addresses are different.

This shows that concatenation creates a **new string object** rather than modifying the existing one.

# Do String Methods Modify the Original String?

Most string methods return a **new string**.

Example:

```python
text = "python"

upper_text = text.upper()
```

The original string remains unchanged.

````python
text = "python"

upper_text = text.upper()

print(text)
print(upper_text)
````

> Although these operations appear to "change" the text, Python actually creates **new strings** each time.
>
> This predictable behavior helps avoid accidental modification of the original data.

# Quick Recap

✔ Strings are **immutable**.

✔ Individual characters cannot be modified.

✔ Most string methods return a **new string**.

✔ Concatenation creates a new string.

✔ Reassignment changes the variable's reference, not the original string.

✔ Immutability improves reliability and memory efficiency.

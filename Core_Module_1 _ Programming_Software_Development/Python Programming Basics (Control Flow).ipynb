# What is Control Flow?

A program normally executes statements from top to bottom.

Sometimes, we want the program to make decisions.

For example:

- If a student passes, display **"Pass"**.
- If the temperature is high, turn on the AC.
- If model accuracy is greater than 95%, save the model.

This process is called **Control Flow** or **Decision Making**.

Control flow allows the program to choose different execution paths based on conditions.

# Program Flow

Without Decision Making

```text
Start

↓

Statement 1

↓

Statement 2

↓

Statement 3

↓

End
```

With Decision Making

```text
          Condition

         /        \

      True       False

       |            |

Statement A   Statement B

       \        /

          Continue
```

# Logical Operators

Logical operators combine multiple conditions.

Python provides three logical operators:

- `and`
- `or`
- `not`

# Logical Operators

| Operator | Meaning |
|-----------|---------|
| `and` | True only if both conditions are True |
| `or` | True if at least one condition is True |
| `not` | Reverses the result |

````python
age = 22
cgpa = 8.5

print(age >= 18 and cgpa >= 8)
````

Both conditions must be `True`.

```python
age >= 18

cgpa >= 8
```

Only then does the final result become `True`.

````python
attendance = 80
medical_certificate = True

print(attendance >= 75 or medical_certificate)
````

Only one condition needs to be `True`.

````python
is_logged_in = False

print(not is_logged_in)
````

`not` reverses the Boolean value.

```text
True → False

False → True
```

# The `if` Statement

The `if` statement executes a block of code **only when a condition is True**.

### Syntax

```python
if condition:
    statement
```

Notice the colon (`:`) after the condition.

The statements inside the `if` block must be properly indented.

````python
age = 20

if age >= 18:
 print("Eligible to Vote")
````

````python
age = 16

if age >= 18:
    print("Eligible to Vote")

print("Program Finished")
````

Since the condition is `False`, the statement inside the `if` block is skipped.

Execution continues with the next statement outside the `if` block.

# The `if-else` Statement

The `if` statement executes code only when a condition is `True`.

But what if we also want to execute another block when the condition is `False`?

For that, Python provides the **`if-else` statement**.

### Syntax

```python
if condition:
    # Executes when condition is True
else:
    # Executes when condition is False
```

Only **one** of the two blocks executes.

````python
age = 17

if age >= 18:
    print("Eligible to Vote")
else:
    print("Not Eligible to Vote")
````

````python
marks = 82

if marks >= 40:
    print("Pass")
else:
    print("Fail")
````

# Flow of Execution

```text
          Condition

         /        \

      True       False

       |            |

      if         else

       \        /

        Continue Program
```

````python
accuracy = 91.5

if accuracy >= 90:
    print("Model Accepted")
else:
    print("Model Needs Improvement")
````

````python
missing_values = 0

if missing_values == 0:
    print("Dataset Ready")
else:
    print("Clean the Dataset")
````

# Combining Multiple Conditions

Logical operators allow us to check multiple conditions together.

Example:

```python
if condition1 and condition2:
```

or

```python
if condition1 or condition2:
```

````python
age = 17
cgpa = 8.6

if age >= 18 and cgpa >= 8:
    print("Eligible for Placement")
else:
    print("Not Eligible")
````

````python
attendance = 70
medical_certificate = True

if attendance >= 75 or medical_certificate:
    print("Allowed to Appear")
else:
    print("Not Allowed")
````

````python
logged_in = False

if not logged_in:
    print("Please Login")
else:
    print("Welcome")
````

# The `if-elif-else` Statement

Sometimes there are more than two possible outcomes.

Instead of writing multiple independent `if` statements,

use `if-elif-else`.

### Syntax

```python
if condition1:
    ...

elif condition2:
    ...

elif condition3:
    ...

else:
    ...
```

Python checks conditions from top to bottom.

As soon as one condition becomes `True`, the remaining conditions are skipped.

````python
marks = 86

if marks >= 90:
    print("Grade A+")

elif marks >= 80:
    print("Grade A")

elif marks >= 70:
    print("Grade B")

elif marks >= 60:
    print("Grade C")

elif marks >= 40:
    print("Grade D")

else:
    print("Fail")
````

````python
temperature = 35

if temperature >= 40:
    print("Very Hot")

elif temperature >= 30:
    print("Hot")

elif temperature >= 20:
    print("Pleasant")

else:
    print("Cold")
````

````python
accuracy = 94.6

if accuracy >= 98:
    print("Excellent")

elif accuracy >= 95:
    print("Very Good")

elif accuracy >= 90:
    print("Good")

elif accuracy >= 80:
    print("Average")

else:
    print("Needs Improvement")
````

# Order of Conditions Matters

Python checks conditions from top to bottom.

Example:

```python
marks = 95

if marks >= 40:
    print("Pass")

elif marks >= 90:
    print("Excellent")
```

Output:

```text
Pass
```

The second condition is never checked because the first condition is already `True`.

Always write **more specific conditions first**.

````python
marks = 95

if marks >= 90:
    print("Excellent")

elif marks >= 40:
    print("Pass")

else:
    print("Fail")
````

# Nested `if`

An `if` statement inside another `if` statement is called a **Nested if**.

Nested `if` statements are useful when one decision depends on another.

````python
age = 22
citizen = True

if age >= 18:
    if citizen:
        print("Eligible to Vote")
````

````python
dataset_size = 15000
missing_values = 0

if dataset_size > 10000:
    if missing_values == 0:
        print("Dataset Ready for Training")
````

````python
accuracy = 96
precision = 94

if accuracy >= 95:
    if precision >= 90:
        print("Deploy the Model")
    else:
        print("Improve Precision")
else:
    print("Train Again")
````

# Multiple Independent `if` Statements

Independent `if` statements are different from `if-elif-else`.

Each `if` statement is evaluated separately.

````python
marks = 95

if marks >= 40:
    print("Pass")

if marks >= 80:
    print("Distinction")

if marks >= 90:
    print("Outstanding")
````

Output:

```text
Pass

Distinction

Outstanding
```

All three conditions are checked independently.

````python
confidence = 0.93

if confidence >= 0.90:
    print("High Confidence Prediction")
else:
    print("Low Confidence Prediction")
````

````python
epochs = 120

if epochs >= 100:
    print("Training Completed")
else:
    print("Continue Training")
````

````python
rows = 8500

if rows >= 50000:
    print("Large Dataset")

elif rows >= 10000:
    print("Medium Dataset")

else:
    print("Small Dataset")
````

### ❌ Mistake 1

Writing multiple `if` statements when only one outcome should occur.

Prefer:

```python
if ...

elif ...

else ...
```

---

### ❌ Mistake 2

Writing conditions in the wrong order.

Always place the **most restrictive** condition first.

---

### ❌ Mistake 3

Forgetting indentation inside nested `if` statements.

---

### ❌ Mistake 4

Using `if` instead of `elif`.

Remember:

- `if-elif-else` → only one block executes.
- Multiple `if` statements → every condition is checked.

✔ Nested `if` statements allow one decision inside another.

✔ Multiple independent `if` statements are different from `if-elif-else`.

# Advanced Decision Making

Python provides several shortcuts and advanced control flow features that make code cleaner and more readable.

In this section, you'll learn:

These features are commonly used in professional Python code.

# Short-hand `if`

If an `if` statement contains only **one statement**, it can be written in a single line.

### Syntax

```python
if condition: statement
```

Use this only for simple statements.

````python
marks = 85

if marks >= 40: print("Pass")
````

````python
accuracy = 96

if accuracy >= 95: print("Deploy Model")
````

Avoid using short-hand `if` when multiple statements are required.

For longer logic, always use proper indentation.

# Ternary Operator

Python provides a compact way to write simple `if-else` statements.

### Syntax

```python
value_if_true if condition else value_if_false
```

This is also called the **Conditional Expression**.

````python
marks = 75

result = "Pass" if marks >= 40 else "Fail"

print(result)
````

````python
age = 17

status = "Adult" if age >= 18 else "Minor"

print(status)
````

````python
accuracy = 92

status = "Accepted" if accuracy >= 90 else "Rejected"

print(status)
````

The ternary operator is best suited for **simple conditions**.

Avoid using it for complex decision-making because readability decreases.

````python
marks = 85

grade = (
    "A"
    if marks >= 80
    else "B"
    if marks >= 60
    else "C"
)

print(grade)
````

Nested ternary operators are valid but can quickly become difficult to read.

Prefer `if-elif-else` for multiple conditions.

# Truthy and Falsy Values

Python automatically treats certain values as either **True** or **False**.

You do not always need to compare values explicitly.

Example:

```python
if data:
```

instead of

```python
if len(data) > 0:
```

# Common Truthy and Falsy Values

## Falsy Values

- `False`
- `None`
- `0`
- `0.0`
- `""` (Empty String)
- `[]` (Empty List)
- `()` (Empty Tuple)
- `{}` (Empty Dictionary)
- `set()` (Empty Set)

Everything else is generally considered **Truthy**.

````python
numbers = []

if numbers:
    print("List has data")
else:
    print("List is Empty")
````

````python
text = True

if text:
    print("String is not Empty")
````

````python
value = 0

if value:
    print("True")
else:
    print("False")
````

# The `pass` Statement

Sometimes you need to write code later but want your program to run now.

Python does not allow empty code blocks.

The `pass` statement acts as a placeholder.

### Syntax

```python
if condition:
    pass
```

````python
marks = 30

if marks >= 40:
    pass

print("Program Continues")
````

````python
def future_function():
    pass
````

`pass` performs no operation.

It simply allows the program to continue without producing an error.

# match-case Statement

Python 3.10 introduced the `match-case` statement.

It works similarly to the **switch-case** statement found in other programming languages.

### Syntax

```python
match variable:

    case value1:
        ...

    case value2:
        ...

    case _:
        ...
```

The underscore (`_`) acts as the default case.

````python
day = 3

match day:
    case 1:
        print("Monday")

    case 2:
        print("Tuesday")

    case 3:
        print("Wednesday")

    case _:
        print("Invalid Day")
````

````python
algorithm = "Random Forest"

match algorithm:

    case "Decision Tree":
        print("Tree-Based Model")

    case "Random Forest":
        print("Ensemble Learning")

    case "SVM":
        print("Support Vector Machine")

    case _:
        print("Unknown Algorithm")
````

Use `match-case` when comparing a variable against many fixed values.

For complex logical conditions, continue using `if-elif-else`.

````python
rows = 25000

status = (
    "Large Dataset"
    if rows >= 20000
    else "Small Dataset"
)

print(status)
````

````python
prediction = True

if prediction:
    print("Prediction Available")
else:
    print("No Prediction")
````

````python
model = "KNN"

match model:

    case "KNN":
        print("Nearest Neighbor")

    case "SVM":
        print("Support Vector Machine")

    case "Naive Bayes":
        print("Probabilistic Model")

    case _:
        print("Model Not Supported")
````

### ❌ Mistake 1

Using the ternary operator for complex logic.

Prefer `if-elif-else` when readability is important.

---

### ❌ Mistake 2

Forgetting the default case in `match-case`.

Always include:

```python
case _:
```

---

### ❌ Mistake 3

Leaving an empty code block.

Incorrect:

```python
if marks > 40:
```

Correct:

```python
if marks > 40:
    pass
```

---

### ❌ Mistake 4

Comparing empty collections unnecessarily.

Instead of:

```python
if len(data) > 0:
```

Use:

```python
if data:
```

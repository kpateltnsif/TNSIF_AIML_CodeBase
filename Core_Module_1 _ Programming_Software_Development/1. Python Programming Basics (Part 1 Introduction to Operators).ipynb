# Python Execution Flow

Have you ever wondered what happens after you run a Python program?

The following diagram explains the execution process.

```text
            Python Program
                   │
                   ▼
        Python Interpreter
                   │
                   ▼
         Bytecode Generation
                   │
                   ▼
      Python Virtual Machine (PVM)
                   │
                   ▼
             Machine Code
                   │
                   ▼
               Output
```

### Step-by-Step Explanation

1. You write Python source code.
2. The Python Interpreter checks for syntax errors.
3. The interpreter converts the code into **bytecode**.
4. The Python Virtual Machine (PVM) executes the bytecode.
5. The operating system converts instructions into machine-level operations.
6. The final output is displayed.

Unlike languages such as C++, Python is an **interpreted language**, allowing faster development and easier debugging.

# Python Syntax

Syntax refers to the **set of rules** that define how Python code should be written.

Every programming language has its own syntax.

One of Python's greatest strengths is its **simple and readable syntax**, making it an excellent language for beginners as well as professionals working in Artificial Intelligence and Data Science.

Unlike many programming languages, Python uses **indentation (spaces)** to define blocks of code instead of braces `{}`.

## Python vs Other Programming Languages

Let's compare a simple "Hello World" program.

### C++

```cpp
#include<iostream>
using namespace std;

int main()
{
    cout << "Hello World";
    return 0;
}
```

### Java

```java
public class Main{
    public static void main(String args[]){
        System.out.println("Hello World");
    }
}
```

### Python

```python
print("Hello World")
```

Notice how much simpler Python is.

This simplicity allows Data Scientists and Machine Learning Engineers to focus more on solving problems rather than writing lengthy code.

# Basic Syntax Rules

Python follows some important syntax rules:

- Statements are executed from top to bottom.
- Python is **case-sensitive**.
- Indentation is mandatory.
- Variables do not require explicit declaration.
- Semicolons (`;`) are optional.
- Parentheses are required for functions.

Understanding these rules will help you avoid many beginner mistakes.

`````python
name = "Alice"

print(name)
`````

### Expected Output

```
Alice
```

Python is **case-sensitive**.

This means:

- `name`
- `Name`
- `NAME`

are considered **three different identifiers**.

`````python
name = "Alice"
Name = "Bob"
NAME = "Charlie"

print(name)
print(Name)
print(NAME)
`````

### Expected Output

```
Alice
Bob
Charlie
```

Even though the words look similar, Python treats them as different variable names.

# Indentation

One of Python's unique features is **indentation**.

Indentation refers to the spaces at the beginning of a line of code.

Most programming languages use curly braces `{}` to define blocks of code.

Python uses indentation instead.

By convention, **4 spaces** are used for one level of indentation.

## Visual Representation

Everything aligned under the `if` statement belongs to that block.

Once indentation ends, Python considers the block complete.

Imagine your kitchen.

```
┌──────────────┐
```

# Definition

A **variable** is a name that refers to an object stored in memory.

Think of it as a label attached to a piece of data.

Unlike many programming languages, Python variables do **not** store the data directly.

Instead, they store a **reference** to the object in memory.

# Variables and Memory

Let's understand what happens behind the scenes.

Suppose we write:

```python
age = 20
```

Python performs several steps internally.

1. Creates the integer object `20`.
2. Stores it in memory.
3. Creates the variable `age`.
4. Makes `age` point to that object.

## Memory Representation

```text
              Memory

        +------------------+
        |        20        |
        +------------------+
                 ▲
                 │
              age
```

Notice that the variable does **not contain** the value.

It simply points to the object stored somewhere in memory.

`````python
age = 20

print(age)
`````

### Expected Output

```
20
```

# Changing a Variable

Variables can point to different objects over time.

Consider the following example.

`````python
age = 20

print(age)

age = 25

print(age)
`````

### Expected Output

```
20
25
```

The variable now points to a **different object**.

Python automatically updates the reference.

## Updated Memory Diagram

Initially

```text
age
 │
 ▼
20
```

After assignment

```text
age
 │
 ▼
25
```

The old object may later be removed by Python's **Garbage Collector** if no variables reference it.

Unlike languages such as C, C++, and Java, Python is a **dynamically typed language**.

This is one of the major reasons why Python is easy to learn and highly productive.

# Static Typing vs Dynamic Typing

Before understanding dynamic typing, let's understand static typing.

### Static Typing (C, C++, Java)

When declaring a variable, you must specify its data type.

Example (C++)

```cpp
int age = 20;
```

Here,

- `int` tells the compiler that `age` can only store integers.

If you later try to store text:

```cpp
age = "Alice";
```

The compiler reports an error.

---

### Dynamic Typing (Python)

Python automatically determines the data type during execution.

```python
age = 20
```

No data type is specified.

Python figures it out automatically.

# Visual Comparison

```text
C++

int age = 20

┌──────────┐
│ Variable │────► Integer Only
└──────────┘


Python

age = 20

┌──────────┐
│ Variable │────► Any Object
└──────────┘
```

A Python variable is not permanently tied to a specific data type.

`````python
age = 20

print(age)
print(type(age))
`````

### Expected Output

```text
20
<class 'int'>
```

The `type()` function tells us the type of the object referenced by the variable.

# Reassigning Variables

Since Python is dynamically typed, the same variable can refer to different types of objects during execution.

`````python
value = 100

print(value)
print(type(value))

value = "Machine Learning"

print(value)
print(type(value))

value = 98.75

print(value)
print(type(value))
`````

### Expected Output

```text
100
<class 'int'>

Machine Learning
<class 'str'>

98.75
<class 'float'>
```

Notice that the variable name remained the same.

Only the object it refers to changed.

# Variables are References

Many beginners believe that variables store values directly.

This is **not** how Python works.

Instead,

Variables store **references** to objects.

Suppose we write

```python
marks = 95
```

Internally,

```text
marks
   │
   ▼
+--------+
|   95   |
+--------+
```

The variable points to an object stored somewhere in memory.

# Object Identity

Every object created in Python has a unique identity.

Python provides the `id()` function to view this identity.

Syntax

```python
id(object)
```

The returned value is an integer representing the memory identity of the object during the program's execution.

`````python
number = 50

print(number)
print(id(number))
`````

Example Output

```text
50
140402169504400
```

The exact number shown on your computer may be different.

That is completely normal.

The identity depends on where Python stores the object in memory.

# Value vs Identity

Two different objects can contain the same value.

Identity tells us **where** the object exists.

Value tells us **what** the object contains.

These are two different concepts.

`````python
a = 100
b = 100

print(a)
print(b)

print(id(a))
print(id(b))
`````

Depending on Python's optimization, both variables may refer to the same object.

For small integers, Python often **reuses existing objects** instead of creating new ones.

This optimization improves memory efficiency.

# Python Memory Model

Let's see what happens internally.

```python
x = 10
```

Python performs the following steps:

1. Creates an integer object (`10`) if it doesn't already exist.
2. Stores it in memory.
3. Creates the variable `x`.
4. Makes `x` reference the object.

```text
Memory

+---------+
|   10    |
+---------+
     ▲
     │
     x
```

Now,

```python
y = x
```

Python does **not** create another integer object.

Instead,

```text
+---------+
|   10    |
+---------+
   ▲   ▲
   │   │
   x   y
```

Both variables refer to the same object.

`````python
x = 10
y = x

print(x)
print(y)

print(id(x))
print(id(y))
`````

Observe the output carefully.

The identities are usually the same because both variables reference the same object.

# Reassigning a Variable

Now suppose we change one variable.

```python
x = 10
y = x

x = 50
```

What happens?

Many beginners think both variables become `50`.

This is incorrect.

Instead,

```text
Before

+---------+
|   10    |
+---------+
 ▲       ▲
 │       │
 x       y


After

+---------+
|   10    |
+---------+
      ▲
      │
      y


+---------+
|   50    |
+---------+
      ▲
      │
      x
```

Only the reference of `x` changes.

`````python
x = 10
y = x

x = 50

print(x)
print(y)

print(id(x))
print(id(y))
`````

# Operators

## Learning Objectives

After completing this section, you will be able to:

- Understand what operators are.
- Perform arithmetic operations.
- Compare values.
- Use logical operators.
- Work with assignment operators.
- Understand membership and identity operators.
- Apply operators in AI & Data Science programs.

---

An **operator** is a symbol that tells Python to perform a specific operation on one or more values.

Example:

```python
5 + 3
```

Here,

- `5` and `3` are **operands**
- `+` is the **operator**

The result is:

```text
8
```

# Types of Operators in Python

Python provides several categories of operators.

| Operator Type | Purpose |
|---------------|---------|
| Arithmetic | Mathematical calculations |
| Comparison | Compare two values |
| Logical | Combine conditions |
| Assignment | Assign values |
| Membership | Check if a value exists |
| Identity | Check object identity |
| Bitwise | Binary operations (Advanced) |

In this notebook, we'll cover everything except Bitwise Operators, which are beyond the scope of Python basics.

# Arithmetic Operators

Arithmetic operators perform mathematical calculations.

| Operator | Meaning | Example |
|----------|---------|---------|
| `+` | Addition | 5 + 2 |
| `-` | Subtraction | 5 - 2 |
| `*` | Multiplication | 5 * 2 |
| `/` | Division | 5 / 2 |
| `//` | Floor Division | 5 // 2 |
| `%` | Modulus | 5 % 2 |
| `**` | Exponent | 5 ** 2 |

`````python
a = 15
b = 4

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
`````

## Understanding Floor Division

Floor Division (`//`) returns the integer quotient after removing the decimal part.

Example:

```python
15 // 4
```

Result:

```text
3
```

Normal Division:

```python
15 / 4
```

Result:

```text
3.75
```

## Understanding Modulus (%)

The modulus operator returns the **remainder**.

Example

```text
15 ÷ 4

Quotient = 3
Remainder = 3
```

Therefore,

```python
15 % 4
```

returns

```text
3
```

Modulus is commonly used to:

- Check if a number is even or odd.
- Check divisibility.
- Create repeating patterns.

`````python
number = 18

print(number % 2)
`````

If the remainder is **0**, the number is even.

Otherwise, it is odd.

# Comparison Operators

Comparison operators compare two values and return either:

- `True`
- `False`

| Operator | Meaning |
|----------|---------|
| `==` | Equal |
| `!=` | Not Equal |
| `>` | Greater Than |
| `<` | Less Than |
| `>=` | Greater Than or Equal |
| `<=` | Less Than or Equal |

`````python
x = 10
y = 20

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
`````

Comparison operators always produce a Boolean value:

```text
True
```

or

```text
False
```

These operators form the foundation of decision-making in Python.

# Logical Operators

Logical operators combine multiple conditions.

| Operator | Meaning |
|----------|---------|
| `and` | Both conditions must be True |
| `or` | At least one condition must be True |
| `not` | Reverses the result |

`````python
age = 22
salary = 40000

print(age > 18 and salary > 30000)
print(age > 25 or salary > 30000)
print(not age > 18)
`````

## Truth Table

### AND

| A | B | Result |
|---|---|--------|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

### OR

| A | B | Result |
|---|---|--------|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |

# Assignment Operators

Assignment operators assign values to variables.

| Operator | Example |
|----------|---------|
| `=` | x = 10 |
| `+=` | x += 5 |
| `-=` | x -= 5 |
| `*=` | x *= 5 |
| `/=` | x /= 5 |
| `//=` | x //= 5 |
| `%=` | x %= 5 |
| `**=` | x **= 5 |

`````python
x = 10

x += 5
print(x)

x *= 2
print(x)

x -= 4
print(x)
`````

# Membership Operators

Membership operators check whether a value exists inside a sequence.

| Operator | Meaning |
|----------|---------|
| `in` | Exists |
| `not in` | Does not exist |

These operators work with:

- Strings
- Lists
- Tuples
- Sets
- Dictionaries

`````python
language = "Python AI"

print("P" in language)
print("z" in language)
print("AI" not in language)
`````

`````python
numbers = [10, 20, 30, 40]

print(20 in numbers)
print(50 in numbers)
`````

# Identity Operators

Identity operators compare whether two variables refer to the **same object**.

| Operator | Meaning |
|----------|---------|
| `is` | Same object |
| `is not` | Different object |

Do not confuse identity with equality.

`````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)

print(a == c)
`````

### Understanding the Difference

```python
a == c
```

Checks whether the **values** are equal.

```python
a is c
```

Checks whether both variables refer to the **same object**.

These are two different concepts.

# Operator Precedence

When an expression contains multiple operators, Python follows a specific order.

A simplified order is:

1. `()`
2. `**`
3. `*`, `/`, `//`, `%`
4. `+`, `-`
5. Comparison Operators
6. `not`
7. `and`
8. `or`

`````python
result = 10 + 5 * 2

print(result)
`````

`````python
result = (10 + 5) * 2

print(result)
`````

Always use parentheses when an expression becomes difficult to read.

Readable code is better than clever code.

# Writing Clean Python Code

Writing code that works is important.

Writing code that is **easy to read, understand, and maintain** is even more important.

Professional Python developers follow a coding standard called **PEP 8**.

PEP stands for **Python Enhancement Proposal**.

PEP 8 provides guidelines for writing consistent and readable Python code.

# Why Follow PEP 8?

Imagine working in a company with 100 developers.

If everyone writes code in their own style, it becomes difficult to:

- Read code
- Find bugs
- Maintain projects
- Collaborate with teammates

Following a common style makes code easier for everyone to understand.

# Variable Naming

Choose variable names that clearly describe the data they store.

### Good Examples

```python
student_name
total_marks
average_salary
learning_rate
batch_size
```

### Poor Examples

```python

b
x
temp
abc
```

Meaningful names make programs easier to read and debug.

`````python
student_name = "Rahul"
marks = 92
average_marks = 89.5

print(student_name)
print(marks)
print(average_marks)
`````

# Naming Conventions

Python uses different naming styles depending on what you are creating.

| Item | Convention | Example |
|------|------------|---------|
| Variable | snake_case | `student_name` |
| Function | snake_case | `calculate_total()` |
| Class | PascalCase | `StudentDetails` |
| Constant | UPPER_CASE | `PI` |
| Module | lowercase | `math_utils.py` |

Throughout this course, we will follow these conventions.

# What is a Set?

A **set** is an unordered collection of **unique** elements.

Unlike lists and tuples:

- Sets do **not** allow duplicate values.
- Sets are **mutable**, meaning elements can be added or removed.
- Sets do **not** support indexing or slicing.

Sets are widely used in data cleaning, duplicate removal, and mathematical operations.

# List vs Tuple vs Set

| Feature | List | Tuple | Set |
|---------|------|-------|-----|
| Ordered | ✅ | ✅ | ❌ |
| Mutable | ✅ | ❌ | ✅ |
| Duplicates | ✅ | ✅ | ❌ |
| Indexing | ✅ | ✅ | ❌ |
| Slicing | ✅ | ✅ | ❌ |
| Uses | General Data | Fixed Data | Unique Data |

# Creating Sets

Sets are created using curly braces `{}`.

### Syntax

```python

```

Duplicate values are automatically removed.

````python
numbers = {10, 20, 20, 30, 40, 30, 50}

print(numbers)
````

The exact order of elements may differ each time because sets are unordered.

# Modifying Sets

Sets are **mutable**, meaning elements can be added or removed after creation.

Python provides several methods to modify sets.

We'll learn:

- `add()`
- `update()`
- `remove()`
- `discard()`
- `pop()`
- `clear()`

# add()

The `add()` method inserts **one element** into a set.

### Syntax

```python
set_name.add(value)
```

If the element already exists, nothing happens.

````python
numbers = {10, 20, 30}

numbers.add(40)

print(numbers)
````

````python
numbers = {10, 20, 30}

numbers.add("Kartik")

print(numbers)
````

Since sets store only unique values, adding an existing element has no effect.

# update()

The `update()` method adds **multiple elements** to a set.

### Syntax

```python
set_name.update(iterable)
```

The iterable can be:

- List
- Tuple
- Set
- String

````python
numbers = {10, 20, 30}

numbers.update([40, 50, 60])

print(numbers)
````

````python
numbers = {10, 20}

numbers.update((20, 40))

print(numbers)
````

````python
numbers = {10, 20}

numbers.update({30, 40})

print(numbers)
````

````python
letters = {"A", "B"}

letters.update("Kartik", " ")

print(letters)
````

# remove()

Removes a specified element.

### Syntax

```python
set_name.remove(value)
```

If the element does not exist, Python raises a `KeyError`.

````python
numbers = {10, 20, 30, 40}

numbers.remove(20)

print(numbers)
````

# discard()

Works similarly to `remove()`.

The difference is:

- `remove()` → Raises an error if the value is absent.
- `discard()` → Does nothing if the value is absent.

````python
numbers = {10, 20, 30}

numbers.discard(50)

print(numbers)
````

# remove() vs discard()

| remove() | discard() |
|-----------|------------|
| Raises `KeyError` | No error |
| Use when the value must exist | Use when you're unsure |

# pop()

The `pop()` method removes **one arbitrary element** from a set.

Since sets are unordered, you cannot predict which element will be removed.

````python
numbers = {10, 20, 30, 40,50,70,80,100, 120 , 140,56,89,43,90,56,34,21}

removed = numbers.pop()

print("Removed :", removed)
print("Remaining :", numbers)
````

Do **not** assume `pop()` removes the first or last element.

The removed element depends on Python's internal implementation.

# clear()

Removes all elements from a set.

The set itself still exists.

````python
numbers = {10, 20, 30}

numbers.clear()

print(numbers)
````

# Set Operations

Sets support powerful mathematical operations.

# Union

Union combines all **unique** elements from both sets.

### Operator

```python
|
```

### Method

```python
union()
```

````python
set1 = {1, 2, 3}

set2 = {3, 4, 5}

print(set1 | set2)
````

````python
set1 = {1, 2, 3}

set2 = {3, 4, 5}

result = set2.union(set1)

print(result)
````

# Intersection

Intersection returns only the **common elements**.

### Operator

```python
&
```

### Method

```python
intersection()
```

````python
set1 = {1, 2, 3, 4}

set2 = {3, 4, 5, 6}

print(set1 & set2)
````

````python
print(set2.intersection(set1))
````

> ## 🧠 AI Example
>
> Suppose:
>
> - Set A = Customers who clicked an advertisement.
> - Set B = Customers who purchased a product.
>
> The **intersection** gives customers who both clicked the ad **and** made a purchase.

#Difference


````python
set1 = {1, 2, 3, 4}

set2 = {3, 4, 5}

print(set1 - set2)
````

````python
print(set2 - set1)
````

````python
result = set1.difference(set2)

print(result)
````

Notice:

```python
A - B
```

is **not** the same as

```python
B - A
```

# Symmetric Difference

Returns elements present in **either set but not both**.

### Operator

```python
^
```

### Method

```python
symmetric_difference()
```

````python
set1 = {1, 2, 3}

set2 = {3, 4, 5}

print(set1 ^ set2)
````

````python
print(set1.symmetric_difference(set2))
````

````python
sentence = [
    "AI",
    "ML",
    "Python",
    "AI",
    "Python",
    "Data"
]

unique_words = set(sentence)

print(unique_words)
````

````python
model1_features = {
    "Age",
    "Salary",
    "Experience"
}

model2_features = {
    "Salary",
    "Experience",
    "Education"
}

print(model1_features & model2_features)
````

# Subset

A set **A** is a subset of set **B** if every element of **A** exists in **B**.

### Methods

```python
issubset()
```

### Operator

```python
<=
```

````python
set1 = {10, 20}

set2 = {10, 20, 30, 40}

print(set1.issubset(set2))
````

````python
set1 = {10, 20}

set2 = {10, 20, 30, 40}

print(set1 <= set2)
````

# Superset

A set **A** is a superset of **B** if it contains every element of **B**.

### Methods

```python
issuperset()
```

### Operator

```python
>=
```

````python
set1 = {10, 20, 30, 40}

set2 = {20, 30}

print(set1.issuperset(set2))

print(set1 >= set2)
````

# Disjoint Sets

Two sets are **disjoint** if they have **no common elements**.

### Method

```python
isdisjoint()
```

````python
set1 = {1, 2, 3}

set2 = {1, 5, 6}

print(set1.isdisjoint(set2))
````

````python
set1 = {1, 2, 3}

set2 = {3, 4, 5}

print(set1.isdisjoint(set2))
````

> ## 🧠 AI Example
>
> Two customer groups can be checked to determine whether they contain any common customers.
>
> If `isdisjoint()` returns `True`, there are no overlapping users.

# Frozen Sets

A **frozenset** is an immutable version of a set.

After creation:

- No elements can be added.
- No elements can be removed.
- No elements can be modified.

### Syntax

```python
frozenset(iterable)
```

````python
numbers = frozenset([10, 20, 30, 40])

print(numbers)
````

````python
numbers = frozenset([10, 20, 30])

# Uncomment the next line to observe the error

# numbers.add(40)
````

Frozen sets are useful when a collection of unique values should never change.

Examples include:

- Fixed feature sets
- Constant labels
- Configuration values

# Traversing a Set

Although sets are unordered, they can still be traversed using a loop.

````python
courses = {
    "Python",
    "Machine Learning",
    "Data Science"
}

for course in courses:
    print(course)
````

````python
courses = {
    "Python",
    "Machine Learning",
    "Data Science"
}

for course in range(len(courses)):
    print(course)
````

The order may differ each time the program runs because sets are unordered.

# Useful Built-in Functions

Several built-in Python functions work with sets.

````python
numbers = {45, 12, 89, 34, 67}

print("Length :", len(numbers))
print("Maximum :", max(numbers))
print("Minimum :", min(numbers))
print("Sum :", sum(numbers))
print("Sorted :", sorted(numbers))
````

`sorted()` returns a **list**, not a set.

````python
students = [
    "Rahul",
    "Aisha",
    "Rahul",
    "John",
    "Aisha"
]

unique_students = set(students)

print(unique_students)
````

````python
python_students = {
    "Rahul",
    "Aisha",
    "John"
}

ml_students = {
    "John",
    "Aisha",
    "Priya"
}

print("Common Students :", python_students & ml_students)
````

````python
required_features = {
    "Age",
    "Salary",
    "Experience"
}

available_features = {
    "Age",
    "Salary",
    "Education"
}

print("Missing Features :",
      required_features - available_features)
````

````python
predictions = [
    0,
    1,
    0,
    1,
    2,
    2,
    1,
    0
]

print(set(predictions))
````

## Frequently Asked Interview Questions

### Q1. Why are sets faster than lists for membership checking?

Because sets are implemented using **hash tables**, making membership checks very efficient.

---

### Q2. What is the difference between a set and a frozenset?

| Set | Frozen Set |
|------|------------|
| Mutable | Immutable |
| Can add/remove elements | Cannot modify |
| Uses `set()` | Uses `frozenset()` |

---

### Q3. Why don't sets support indexing?

Because sets are unordered collections and do not maintain element positions.

---

### Q4. When should you use a set?

Use a set when:

- Duplicate values should be removed.
- Fast membership checking is required.
- Mathematical set operations are needed.

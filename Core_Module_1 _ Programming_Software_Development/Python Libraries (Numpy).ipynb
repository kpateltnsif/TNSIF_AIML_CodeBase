# What is NumPy?

**NumPy** stands for **Numerical Python**.

It is a Python library designed for efficient numerical computation.

The central object in NumPy is the:

```python
ndarray
```

which stands for:

> N-dimensional array

A NumPy array can contain numerical data arranged in one or more dimensions.

Example:

```text
1D Array

[10 20 30 40 50]


2D Array

[[10 20 30]
 [40 50 60]]


3D Array

[
    [
        [1 2]
        [3 4]
    ],

    [
        [5 6]
        [7 8]
    ]
]
```

NumPy arrays are extremely important because Machine Learning algorithms work heavily with numerical arrays and matrices.

# Importing NumPy

NumPy must be imported before using it.

The standard convention is:

```python

```

`np` is an alias commonly used for NumPy.

This allows us to write:

```python

np.mean()
np.sum()
```

instead of:

```python

numpy.mean()
numpy.sum()
```

`````python
import numpy as np
`````

`````python
print(np.__version__)
`````

# Note

`np` is only an alias.

You could technically write:

```python
import numpy
```

and then:

```python

```

However, `np` is the standard convention used throughout Data Science and Machine Learning code.

# Python Lists vs NumPy Arrays

Before understanding NumPy arrays, let's see the difference between a normal Python list and a NumPy array.

Python list:

```python

```

NumPy array:

```python

```

Both can store multiple values.

However, NumPy arrays are specifically optimized for numerical computation.

`````python
python_list = [10, 20, 30, 40]

numpy_array = np.array([10, 20, 30, 40])

print(python_list)
print(numpy_array)
`````

# `astype()`

`astype()` creates an array with a different data type.

### Syntax

```python
array.astype(data_type)
```

This is frequently used when preparing numerical data for Machine Learning.

`````python
numbers = np.array([10, 20, 30, 40])

float_numbers = numbers.astype(float)

print(numbers)
print(float_numbers)

print(numbers.dtype)
print(float_numbers.dtype)
`````

`````python
values = np.array([10.5, 20.8, 30.2])

integer_values = values.astype(int)

print(integer_values)
`````

> **Important**
>
> Converting floating-point values to integers removes the decimal portion.
>
> For example:
>
> ```text
> 10.8 → 10
> 20.9 → 20
> ```
>
> It does not round the values.

# Basic Array Operations

One of the major advantages of NumPy is that mathematical operations can be performed directly on arrays.

For example:

```python
array * 2
```

applies the multiplication to every element.

`````python
marks = np.array([70, 75, 80, 85, 90])

print(marks + 5)
`````

`````python
marks = np.array([70, 75, 80, 85, 90])

print(marks * 2)
`````

`````python
salary = np.array([30000, 40000, 50000])

print(salary * 1.10)
`````

This is called **vectorized computation**.

Instead of manually writing:

```python
salary[0] * 1.10
salary[1] * 1.10
salary[2] * 1.10
```

NumPy performs the operation on the entire array.

# `arange()` vs `linspace()`

### `arange()`

Specify the **step size**.

```python
np.arange(0, 10, 2)
```

Result:

```text
0 2 4 6 8
```

### `linspace()`

Specify the **number of values**.

```python
np.linspace(0, 10, 5)
```

Result:

```text
0 2.5 5 7.5 10
```

### Quick Rule

Use:

```text
arange → I know the step size

linspace → I know how many values I need
```

# Broadcasting

**Broadcasting** is one of the most important NumPy concepts.

Broadcasting allows NumPy to perform operations between arrays with different but compatible shapes.

For example:

```python
numbers = np.array([10, 20, 30])

numbers + 5
```

The scalar `5` is effectively applied to every element.

Conceptually:

```text
[10, 20, 30]
+
[ 5,  5,  5]

↓

[15, 25, 35]
```

NumPy performs this without requiring us to manually create the second array.

`````python
numbers = np.array([10, 20, 30, 40])

print(numbers + 10)
`````

`````python
numbers = np.array([10, 20, 30, 40])

print(numbers * 0.1)
`````

This is the simplest form of broadcasting:

```text
Array + Scalar
```

The scalar is applied to every element.

#Broadcasting with 2D Arrays

`````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(data + 5)
`````

`````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(data * 2)
`````

Broadcasting can also occur between a 2D array and a compatible 1D array.

Example:

```text
2D array

[[10 20 30]
 [40 50 60]]

+

1D array

[1 2 3]

=

[[11 22 33]
 [41 52 63]]
```

The 1D array is applied across each row.

`````python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

addition = np.array([1, 2, 3])

print(data + addition)
`````

# Simple Min-Max Scaling

A common numerical transformation is:

```text
x_scaled = (x - min) / (max - min)
```

This transforms values approximately into the range:

```text
0 → 1
```

We can implement the mathematics directly with NumPy.

`````python
values = np.array([10, 20, 30, 40, 50])

minimum = values.min()
maximum = values.max()

scaled = (values - minimum) / (maximum - minimum)

print(scaled)
`````

Notice that NumPy performs:

```python
values - minimum
```

on the entire array.

No explicit loop is required.

This is an example of **vectorized computation + broadcasting**.

`````python
living_area = real_data[:, 1].astype(float)

minimum = living_area.min()
maximum = living_area.max()

living_area_scaled = (
    (living_area - minimum) /
    (maximum - minimum)
)

print(living_area_scaled[:10])
`````

Now the `GrLivArea` feature has been transformed to approximately:

```text
0 → minimum value
1 → maximum value
```

This is one of the basic mathematical ideas behind feature scaling.

# Important Rule

The total number of elements must remain unchanged.

For example:

```python
data.size
```

is:

```text
6
```

Therefore:

```python
data.reshape(2, 3)
```

is valid because:

```text
2 × 3 = 6
```

But:

```python
data.reshape(4, 2)
```

is invalid because:

```text
4 × 2 = 8
```

while the original array has only 6 elements.

`````python
data = np.array([10, 20, 30, 40, 50, 60])

print("Total elements:", data.size)

reshaped = data.reshape(2, 3)

print(reshaped)
`````

`flatten()` returns a **copy** of the data.

Therefore, modifying the flattened array does not modify the original array.

`````python
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = data.flatten()

flat[0] = 999

print("Flattened:", flat)
print("Original:")
print(data)
`````

# `ravel()`

`ravel()` also converts an array into 1D.

```python
array.ravel()
```

However, unlike `flatten()`, `ravel()` may return a **view** of the original array when possible.

Therefore, it can be more memory-efficient.

`````python
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = data.ravel()

print(flat)
`````

`````python
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = data.ravel()

flat[0] = 999

print("Ravel:", flat)
print("Original:")
print(data)
`````

## Quick Difference

```text
flatten()
    ↓
Always returns a copy

ravel()
    ↓
Returns a view when possible
```

Use `flatten()` when you explicitly need an independent copy.

Use `ravel()` when memory efficiency matters and a view is acceptable.

# 1. Convolution

Convolution is the fundamental operation in a CNN.

A small matrix called a **filter** or **kernel** moves across the image.

At every position:

1. The filter overlaps part of the image.
2. Corresponding values are multiplied.
3. The multiplied values are added.
4. The result becomes one value in the output.

The complete output is called a **Feature Map**.

### Concept

Image + Filter

↓

Element-wise Multiplication

↓

Summation

↓

Feature Map

## What Does a Filter Detect?

Different filters can learn different visual patterns.

Early CNN layers may learn:

- Vertical edges
- Horizontal edges
- Diagonal edges
- Corners

Deeper layers can learn:

- Textures
- Shapes
- Object parts
- Complex patterns

The filters are not manually designed in a trained CNN.

Their values are learned during training.

## Kernel Size

Kernel size determines the spatial size of the filter.

Common choices:

- 3 × 3

A 3 × 3 kernel examines a small local region of the image at a time.

## 2. Stride

Stride determines how far the filter moves after each convolution.

### Stride = 1

The filter moves one pixel at a time.

### Stride = 2

The filter moves two pixels at a time.

Increasing the stride generally reduces the spatial dimensions of the output.

````python
conv_stride_1 = layers.Conv2D(
    filters=32,
    kernel_size=3,
    strides=1,
    activation="relu"
)

conv_stride_2 = layers.Conv2D(
    filters=32,
    kernel_size=3,
    strides=2,
    activation="relu"
)

print("Stride 1:", conv_stride_1)
print("Stride 2:", conv_stride_2)
````

## 3. Padding

Padding adds extra pixels around the boundary of an image.

Two common options are:

### Valid Padding

No padding is added.

The output becomes smaller.

### Same Padding

Padding is added so that the spatial dimensions are preserved when stride = 1.

Example:

Input:

28 × 28

3 × 3 kernel

Stride = 1

With `padding="same"`:

Output:

28 × 28

````python
conv_valid = layers.Conv2D(
    32,
    3,
    padding="valid",
    activation="relu"
)

conv_same = layers.Conv2D(
    32,
    3,
    padding="same",
    activation="relu"
)

print("Valid padding:", conv_valid)
print("Same padding:", conv_same)
````

## 4. ReLU Activation

CNNs commonly use ReLU after convolution.

### Formula

ReLU(x) = max(0, x)

Therefore:

- Negative values → 0
- Positive values → unchanged

ReLU introduces non-linearity, allowing the network to learn complex patterns.

# 5. Pooling

Pooling reduces the spatial dimensions of feature maps.

The most common type is **Max Pooling**.

### Max Pooling

A window moves across the feature map and selects the maximum value.

For example:

[1  3]
[2  4]

Max = 4

Pooling helps:

- Reduce computation
- Reduce spatial dimensions
- Provide some translation tolerance
- Retain strong features

````python
pooling = layers.MaxPooling2D(
    pool_size=(2, 2)
)

print(pooling)
````

## Max Pooling Example

Input:

4 × 4

Pooling window:

2 × 2

Stride:

2

Output:

2 × 2

The spatial dimensions are reduced while important high-activation features are retained.

The convolutional layers perform feature extraction.

The dense layers perform classification using the extracted features.

````python
import tensorflow as tf
from tensorflow import keras

cnn_model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(32, 3, activation="relu"),
    layers.MaxPooling2D(2),

    layers.Conv2D(64, 3, activation="relu"),
    layers.MaxPooling2D(2),

    layers.Flatten(),

    layers.Dense(128, activation="relu"),

    layers.Dense(10, activation="softmax")
])

cnn_model.summary()
````

## Understanding the Architecture

### MaxPooling2D

### Conv2D(64)

Learns more complex features using 64 filters.

### Flatten

Converts the feature maps into a one-dimensional vector.

### Dense(128)

Combines extracted features for classification.

### Dense(10)

Produces probabilities for 10 classes using Softmax.

````python
from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
````

## 7. Image Preprocessing

CNNs expect images in a suitable numerical format.

We will:

1. Normalize pixel values
2. Add the channel dimension
3. Prepare the labels

````python
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

X_train = np.expand_dims(X_train, axis=-1)
X_test = np.expand_dims(X_test, axis=-1)

print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
````

````python
y_train = keras.utils.to_categorical(
    y_train,
    10
)

y_test = keras.utils.to_categorical(
    y_test,
    10
)

print("Labels converted to one-hot encoding.")
````

````python
cnn_model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),

    layers.Conv2D(
        32,
        kernel_size=3,
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(pool_size=2),

    layers.Conv2D(
        64,
        kernel_size=3,
        activation="relu",
        padding="same"
    ),

    layers.MaxPooling2D(pool_size=2),

    layers.Conv2D(
        128,
        kernel_size=3,
        activation="relu",
        padding="same"
    ),

    layers.Flatten(),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dropout(0.5),

    layers.Dense(
        10,
        activation="softmax"
    )
])

cnn_model.summary()
````

## 8. Dropout

Dropout is a regularization technique.

During training, randomly selected neurons are temporarily ignored.

This prevents the network from relying too heavily on particular neurons.

### Purpose

Dropout helps reduce **overfitting**.

Example:

# 10. VGG

VGG is a convolutional neural network architecture developed by the Visual Geometry Group.

VGG uses a simple and consistent architecture based heavily on:

- 3 × 3 convolution filters
- ReLU activation
- Max pooling
- Fully connected layers

Common versions include:

- VGG16
- VGG19

VGG16 contains 16 weight layers.

````python
from tensorflow.keras.applications import VGG16

vgg_base = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

vgg_base.summary()
````

# 11. ResNet

ResNet stands for **Residual Network**.

A major idea introduced by ResNet is the **Residual Connection**.

Instead of forcing every layer to learn a complete transformation, the network can learn a residual mapping.

### Residual Block

Input

↘

Layers → Transformation

↗

Shortcut Connection

↓

Addition

↓

Output

These shortcut connections help very deep networks train more effectively.

````python
from tensorflow.keras.applications import ResNet50

resnet_base = ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

resnet_base.summary()
````

````python
resnet_base.trainable = False

resnet_model = keras.Sequential([
    resnet_base,

    layers.GlobalAveragePooling2D(),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dropout(0.5),

    layers.Dense(
        10,
        activation="softmax"
    )
])

resnet_model.summary()
````

## 12. VGG vs ResNet

| VGG | ResNet |
|---|---|
| Simple sequential architecture | Uses residual connections |
| Mainly 3 × 3 convolutions | Uses residual blocks |
| Large number of parameters | More efficient for deep networks |
| Computationally expensive | Designed for very deep networks |
| Easy to understand | More sophisticated architecture |

### Key Idea

**VGG** → Simple and sequential CNN architecture.

**ResNet** → Deep CNN architecture using shortcut/residual connections.

## 13. Fine-Tuning

Initially, we freeze the pretrained layers.

The new classification layers are trained first.

After that, some of the deeper pretrained layers can be unfrozen and trained with a very small learning rate.

This process is called **fine-tuning**.

### Transfer Learning Strategy

1. Load pretrained model
2. Remove original classifier
3. Freeze pretrained layers
4. Add new classifier
5. Train new classifier
6. Unfreeze selected layers
7. Use a small learning rate
8. Continue training

````python
resnet_base.trainable = True

for layer in resnet_base.layers[:-20]:
    layer.trainable = False

print("Last 20 ResNet layers are available for fine-tuning.")
````

## YOLO Annotation Format

Each image has a corresponding `.txt` file.

A YOLO annotation has:

class_id x_center y_center width height

Example:

0 0.52 0.48 0.35 0.25

All bounding-box coordinates are normalized between 0 and 1.

We will convert them into:

x_min
y_min
x_max
y_max

for our CNN model.

#Load Dataset

````python
IMG_SIZE = 224

image_paths = []
boxes = []
classes = []

for filename in os.listdir(train_image_dir):

    if not filename.lower().endswith(
        (".jpg", ".jpeg", ".png")
    ):
        continue

    image_path = os.path.join(
        train_image_dir,
        filename
    )

    label_name = os.path.splitext(filename)[0] + ".txt"

    label_path = os.path.join(
        train_label_dir,
        label_name
    )

    if not os.path.exists(label_path):
        continue

    with open(label_path, "r") as file:
        lines = file.readlines()

    if len(lines) == 0:
        continue

    # Use the first object for this beginner example
    values = lines[0].strip().split()

    class_id = int(values[0])

    x_center = float(values[1])
    y_center = float(values[2])
    width = float(values[3])
    height = float(values[4])

    x_min = x_center - width / 2
    y_min = y_center - height / 2
    x_max = x_center + width / 2
    y_max = y_center + height / 2

    image_paths.append(image_path)

    boxes.append([
        x_min,
        y_min,
        x_max,
        y_max
    ])

    classes.append(class_id)

print("Images loaded:", len(image_paths))
````

#Create DataFrame

````python
data = pd.DataFrame({
    "image_path": image_paths,
    "x_min": [box[0] for box in boxes],
    "y_min": [box[1] for box in boxes],
    "x_max": [box[2] for box in boxes],
    "y_max": [box[3] for box in boxes],
    "class": classes
})

data.head()
````

#Visualize Bounding Boxes

````python
def show_image_with_box(row):

    image = cv2.imread(row["image_path"])
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    height, width = image.shape[:2]

    x1 = int(row["x_min"] * width)
    y1 = int(row["y_min"] * height)

    x2 = int(row["x_max"] * width)
    y2 = int(row["y_max"] * height)

    cv2.rectangle(
        image,
        (x1, y1),
        (x2, y2),
        (255, 0, 0),
        3
    )

    plt.figure(figsize=(8, 6))
    plt.imshow(image)
    plt.axis("off")
    plt.show()
````

````python
show_image_with_box(data.iloc[0])
````

````python
def load_image(path):

    image = cv2.imread(path)

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    image = cv2.resize(
        image,
        (IMG_SIZE, IMG_SIZE)
    )

    image = image.astype("float32") / 255.0

    return image
````

````python
X = np.array([
    load_image(path)
    for path in data["image_path"]
])

y_boxes = data[
    [
        "x_min",
        "y_min",
        "x_max",
        "y_max"
    ]
].values.astype("float32")

y_classes = data["class"].values

print("Image shape:", X.shape)
print("Bounding box shape:", y_boxes.shape)
print("Class shape:", y_classes.shape)
````

````python
(
    X_train,
    X_test,
    y_box_train,
    y_box_test,
    y_class_train,
    y_class_test
) = train_test_split(
    X,
    y_boxes,
    y_classes,
    test_size=0.2,
    random_state=42
)

print("Training images:", len(X_train))
print("Testing images:", len(X_test))
````

# CNN Object Detection Model

Our CNN will have two output heads.

### 1. Classification Head

Predicts the object class.

### 2. Bounding Box Head

Predicts:

- x_min
- y_min
- x_max
- y_max

### Architecture

Image

↓

Convolution

↓

ReLU

↓

Pooling

Convolution

ReLU

↓

Pooling

↓

Convolution

↓

↓

#Build CNN

````python
inputs = keras.Input(
    shape=(IMG_SIZE, IMG_SIZE, 3)
)

x = layers.Conv2D(
    32,
    3,
    activation="relu"
)(inputs)

x = layers.MaxPooling2D(2)(x)

x = layers.Conv2D(
    64,
    3,
    activation="relu"
)(x)

x = layers.MaxPooling2D(2)(x)

x = layers.Conv2D(
    128,
    3,
    activation="relu"
)(x)

x = layers.MaxPooling2D(2)(x)

x = layers.Conv2D(
    256,
    3,
    activation="relu"
)(x)

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dense(
    128,
    activation="relu"
)(x)

# Classification output
class_output = layers.Dense(
    1,
    activation="sigmoid",
    name="class_output"
)(x)

# Bounding box output
box_output = layers.Dense(
    4,
    activation="sigmoid",
    name="box_output"
)(x)

model = keras.Model(
    inputs=inputs,
    outputs=[
        class_output,
        box_output
    ]
)

model.summary()
````

## Loss Functions

Our model has two outputs, therefore we use two losses.

### Classification Loss

Binary Cross-Entropy.

### Bounding Box Loss

Mean Squared Error.

The total training loss is calculated from both outputs.

````python
model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss={
        "class_output": "binary_crossentropy",
        "box_output": "mse"
    },

    metrics={
        "class_output": ["accuracy"],
        "box_output": ["mae"]
    }
)

print("Model compiled successfully.")
````

#Train CNN

````python
history = model.fit(
    X_train,

    {
        "class_output": y_class_train,
        "box_output": y_box_train
    },

    validation_split=0.2,

    epochs=15,

    batch_size=32,

    verbose=1
)
````

#Evaluate

````python
results = model.evaluate(
    X_test,

    {
        "class_output": y_class_test,
        "box_output": y_box_test
    },

    verbose=1
)

print(results)
````

# Intersection over Union (IoU)

IoU measures how much the predicted bounding box overlaps with the actual bounding box.

### Formula

IoU = Area of Intersection / Area of Union

### Interpretation

IoU = 1

Perfect overlap.

IoU = 0

No overlap.

Higher IoU means a better bounding-box prediction.

IoU is one of the fundamental metrics used in object detection.

````python
def calculate_iou(actual, predicted):

    x1 = max(actual[0], predicted[0])
    y1 = max(actual[1], predicted[1])

    x2 = min(actual[2], predicted[2])
    y2 = min(actual[3], predicted[3])

    intersection_width = max(
        0,
        x2 - x1
    )

    intersection_height = max(
        0,
        y2 - y1
    )

    intersection = (
        intersection_width *
        intersection_height
    )

    actual_area = (
        (actual[2] - actual[0]) *
        (actual[3] - actual[1])
    )

    predicted_area = (
        (predicted[2] - predicted[0]) *
        (predicted[3] - predicted[1])
    )

    union = (
        actual_area +
        predicted_area -
        intersection
    )

    if union == 0:
        return 0

    return intersection / union
````

#Calculate IoU

````python
ious = []

for actual, predicted in zip(
    y_box_test,
    box_predictions
):

    iou = calculate_iou(
        actual,
        predicted
    )

    ious.append(iou)

ious = np.array(ious)

print("Mean IoU:", ious.mean())
````

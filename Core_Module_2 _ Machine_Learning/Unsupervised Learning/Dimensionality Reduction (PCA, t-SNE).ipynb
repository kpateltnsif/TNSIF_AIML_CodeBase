## Feature Preparation

PCA and t-SNE work with numerical features.

We will separate the target column and use only the numerical features for dimensionality reduction.

```python
target_column = df.columns[-1]

X = df.drop(columns=[target_column])
y = df[target_column]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Features scaled successfully.")
```

# Principal Component Analysis (PCA)

PCA transforms the original features into a smaller number of new features called **Principal Components**.

The first principal component captures the maximum possible variance.

The second captures the next highest variance while being independent of the first.

### Main Uses

- Reduce dimensions
- Remove redundancy
- Visualize high-dimensional data
- Preserve important variance

```python
pca = PCA()

X_pca = pca.fit_transform(X_scaled)

print("Original dimensions:", X_scaled.shape[1])
print("PCA dimensions:", X_pca.shape[1])
```

```python
explained_variance = pca.explained_variance_ratio_

explained_variance
```

```python
cumulative_variance = np.cumsum(explained_variance)

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Explained Variance")

plt.grid(True)
plt.show()
```

## Selecting Principal Components

We can select enough components to preserve a desired amount of variance.

For example, 95% explained variance means that the selected components retain approximately 95% of the information represented by the original features.

```python
pca_95 = PCA(n_components=0.95)

X_pca_95 = pca_95.fit_transform(X_scaled)

print("Original features:", X_scaled.shape[1])
print("Selected components:", X_pca_95.shape[1])
```

## PCA 2D Visualization

For visualization, we reduce the dataset to two principal components.

```python
pca_2d = PCA(n_components=2)

X_pca_2d = pca_2d.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    X_pca_2d,
    columns=["PC1", "PC2"]
)

pca_df["Target"] = y.values

pca_df.head()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Target",
    palette="tab10",
    legend=False
)

plt.title("PCA — 2D Visualization")
plt.show()
```

# t-SNE

t-SNE (t-Distributed Stochastic Neighbor Embedding) is a nonlinear dimensionality-reduction technique.

It is particularly useful for visualizing high-dimensional data.

Unlike PCA, t-SNE focuses strongly on preserving **local relationships** between data points.

### PCA vs t-SNE

| PCA | t-SNE |
|---|---|
| Linear | Nonlinear |
| Preserves variance | Preserves local structure |
| Fast | Computationally expensive |
| More reproducible | Results can vary |
| Useful for reduction | Mainly useful for visualization |

```python
tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30
)

X_tsne = tsne.fit_transform(X_scaled)

print("t-SNE transformation completed.")
```

```python
tsne_df = pd.DataFrame(
    X_tsne,
    columns=["TSNE1", "TSNE2"]
)

tsne_df["Target"] = y.values

tsne_df.head()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=tsne_df,
    x="TSNE1",
    y="TSNE2",
    hue="Target",
    palette="tab10",
    legend=False
)

plt.title("t-SNE — 2D Visualization")
plt.show()
```
